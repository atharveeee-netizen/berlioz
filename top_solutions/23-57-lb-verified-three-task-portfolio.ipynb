**What is ours.**

1. *The Schur dqi, rebuilt and 9 qubits narrower* (section 3).
   The decoder is jday96314's; we rebuilt it from their code and took its peak from 892 to 883 qubits.
   That is −1.74 % on the dqi cost, which pays qubits squared.
   v38 takes one more off (882): a multi-target product fans out through its own first target instead of a scratch wire, at zero extra Toffolis, and that reverse synthesis is spliced under Özer's Bézout middle block.
2. *An LLM-iterated gauge for the chemistry factor* (section 4).
   It is an orthogonal rotation of the factor that lowers λ, found by Gemini iterating against the official verifier.
   It is applied to the cheapest chemistry whenever it verifies cheaper.
3. *A verified portfolio that stays current* (section 2).
   Every candidate is checked offline with the official verifiers and matched here by hash; the cheapest PASS per task goes in.
   Only new circuits, such as the gauged chemistry, are verified again here, so the notebook runs in a few minutes.
4. *The chemistry, 14.2 % cheaper than any public one.* rustad's factorization has half the weights of the 440-qubit one; mrmeows's angle compiler had only been run on his own factorization. Together, with four cached monomials erased and recomputed mid-circuit so that 4 words per circuit fit the 128-qubit angle cap: 423 qubits, official verifier PASS at 4.2597 × 10⁸, 4.2458 × 10⁸ with our gauge.
5. *Measurements of where the remaining points are* (section 5).

Everything else is other people's work, credited at the end.
Earlier work (sections 6 to 8: a cost anatomy, our 93-qubit invsqrt and a Berlekamp–Massey dqi port) is kept below; the portfolio has moved past it.
The version history is in the changelog at the end.

## What we tried that did not work

Every row below is a measured null: a real build or a logged profiling run, not a guess.
Source docs are in `docs/` of this repo (`HAILMARY_2026-10-27.md`, `DQI_J5_WIDTH.md`, `DQI_PEAK_2026-10-27.md`, `CHEM_IDEAS_2026-10-27.md`; all of them in one table: `QC_NULLS_2026-10-27.md`).

| idea | what we measured | result |
|---|---|---|
| 888-qubit dqi build (drop one qubit from the 889-qubit dqi) | Official verifier cost 4.7590e11 vs 4.4860e11 for the shipped 889-qubit build. | Worse. Rejected, kept under a `rejected_` filename. |
| Chemistry address-basis split (linear change of basis before the angle lookup) | Best layout found cuts Toffolis to 266 per circuit, but needs 169 qubits against the 128-qubit cap. | Not buildable. |
| Dirty-wire AND on the invsqrt row-opening gate | A cheap cover needs only 23 extra Toffolis, under the 26.5-Toffoli break-even budget, but it is not correct: a valid version needs a restore CCX before each read cluster, which costs far more than 26.5 Toffolis. | Net loss. |
| SelectCopy QROM (halve the QROM cost) | Chemistry's QROAM cost is set by the verifier's own formula, not by our circuit. | No lever. Cannot apply here. |
| arXiv 2510.10967 reference DQI design (Khattar, Shutty, Gidney et al.) | Reference design costs 5.72e6 Toffolis (EEA method). | Nothing past what we already ship. |
| GF(2^12) multiplier below rank 42 | No new rank bound found. arXiv 2603.07280 proves lower bounds only, no construction. | No gain. |
| Jet-flag-only 888-qubit dqi (drop the jet completion flag) | Dropping the flag adds 2 value errors on identical inputs, R 1.19126 to 1.20139, +0.85% R. One qubit is worth 0.225% R. | Net loss, about 4x over budget. |
| Peak shaving on Özer's 877-qubit dqi (reorder stages, recompute, borrow idle qubits) | With cost ~ Q²·T one qubit is worth about 1,100 Toffolis. The 877 peak is hit in 2,046 windows over about 684 kernel calls; 0 known-value qubits to borrow at the 1,734 jet windows; the cheapest narrowing costs at least 23 to 42 Toffolis per window. | Nothing under the exchange rate; not built. |
| One chemistry group fewer (R 10 → 9) or fewer copies | Özer's v9 chemistry already sits on both gates: error₂ 19.998 mHa (gate 20), E_gap 3.9151 (floor 3.915); marculera measured nine merged groups at 22.1 mHa after rounding. | Closed before a run. |
| Rounding the chemistry angles for the circuit, or reordering the Givens tree | On Özer's words the mixed (low × high) ANF matrix of every one of the 525 output bits has full GF(2) rank 15, like a random matrix; his 321 Toffolis per circuit sit on the 262-product rank bound plus the caches. Column-pair XORs are 0.95 to 1.0× as complex as single columns. | No structure to remove. |
| GPTQ / OBQ-style rounding of the chemistry angles (inverse-Hessian correction inside each row) | Same 15.26 mHa continuous start and 10-stage schedule: GPTQ + polish 47.04 mHa, plain ±1 polish 46.86, nearest 50.66; after each stage's refit the arms are within 0.5 mHa. | The refit absorbs it; the freeze schedule decides, not the rounding rule. |
| A different Givens tree for the chemistry angles (shape or orbital labels) | Same continuous vectors (Rustad C46 refit) re-encoded exactly on 18 other trees, every angle rounded to 8 bits, from a start shifted off every grid: the starter tree rounds to 219.4 mHa; binary heap +7.8 %, the starter shape with relabelled orbitals +17-29 %, spanning trees on the vectors' correlations +33-43 %, chain or star +79-87 %. | The starter tree is already the best tree tried; its own orbital numbering is worth about 20 %. |
| invsqrt at 62 qubits | manwithacat's *Is the 63-qubit inverse square root optimal?*: the shipped table rounding is the only one that passes in a full search of the nine binding segments; the 44 angles are at the LP optimum. | Closed. |
| Our gauge on Özer's v9 chemistry | 3.551868 × 10⁸ → 3.551853 × 10⁸. | −0.0004 %: his λ loop leaves almost nothing; shipped only as a tie-breaker. |
| LLM evolution of the gauge optimiser (ShinkaEvolve, local Qwen 30B) | 0 improvements in 15 generations. | Null (Gemini in a hand loop improved on its first try). |
| 7-bit quantization-aware retraining of the chemistry factorization (brot 7: the cost-model cliff below) | Measured by [wanxiaolong](https://www.kaggle.com/competitions/google-verifiable-quantum-circuit-compilation/discussion/753687) from somejoint's v2 (8-bit, R8/B143/C60): 7-bit one-body error₁ 0.0176 (Huffman-style Givens tree + annealing encoder; 0.0191 on the original tree), but 7-bit two-body QAT by local search plateaus at error₂ ≈ 0.11 (~130k accepted ±1 word flips with branch-and-bound and w/W refit; 0.13 on the original tree). STE+Adam, soft-to-hard quantization and progressive freezing also failed. An 8-bit control with the same optimizer cannot recover 0.02 from a perturbed known solution. | Above the gate so far. The control says this is an optimizer limit, not a proof that 7-bit is infeasible. |

**Takeaways.** Every width cut we tried on dqi or invsqrt costs more error or more Toffolis than the qubit is worth.
The verifier's own formulas set hard floors we cannot compile around: the QROAM cost formula, and the dqi R floor at zero errors.

## 1. The competition on one page

Every task asks for a **kickmix circuit**: reversible classical gates (X, CX, Toffoli), phase gates, and
measurement-based uncomputation. Such a circuit maps every computational-basis input to one basis output times a phase,
so a classical simulator can run it on random inputs very fast and check it exactly. That is what makes the
competition verifiable: the verifier seeds its test inputs from a hash of your circuit file, so there is no hidden test set.

| task | what the circuit does | cost the verifier reports | reference |
|---|---|---|---|
| **invsqrt** | phases a 22-bit input x by β/√x (pairwise Coulomb energy in fusion simulations) | qubits × (Toffolis + T/4) | 260 qubits, 7,988 |
| **dqi** | decodes a Reed–Solomon syndrome over GF(2¹²) and fixes a sign (decoded quantum interferometry) | R(ε) · qubits² · Toffolis | 1,893 qubits, 5.73 M Toffolis |
| **dfthcsos** | loads the rotation angles of a factorised FeMoCo Hamiltonian (chemistry) with QROM circuits | λ_eff · Toffolis · qubits | 730 qubits, 93 k Toffolis |

**Score.** For each task, *reference cost ÷ your cost*; the leaderboard shows the **geometric mean** of the three.
Submitting the reference circuits scores 1.0. Because it is a geometric mean, doubling the ratio of your *weakest*
task is worth exactly as much as doubling your best one, and a task you leave at 1.0 caps you hard.

```python
ROWS = r'<rect x="([-\d.e]+)" y="([-\d.e]+)" width="([-\d.e]+)" height="10" stroke="black" fill="red"[^>]*/>\s*<text[^>]*>([^<]*)</text>'
def anatomy(path):
    '''Stages of a kickmix circuit from its own flame chart: red bars = tagged stages (width = share), y = nesting.'''
    c = km.Circuit.from_file(str(path)); st = c.operation_statistics(); svg = c.flame_chart_svg()
    full = float(re.search(r'viewBox="[-\d.]+ [-\d.]+ ([\d.]+)', svg).group(1)) - 200
    bars = [(float(y), l.strip(), float(x), float(w)) for x, y, w, l in re.findall(ROWS, svg)]
    top = min((y for y, *_ in bars), default=0)
    stages = [dict(depth=round((y - top) / 10), stage=l, start=x / full, share=w / full) for y, l, x, w in bars]
    return dict(qubits=c.num_qubits, toffoli=st.num_toffoli_like, instructions=st.num_instructions, stages=stages, svg=svg)
def top_table(a, name):
    d0 = sorted((s for s in a['stages'] if s['depth'] == 0), key=lambda s: s['start'])
    rows = [(s['stage'], f"{100 * s['share']:.1f}%", round(s['share'] * a['toffoli'])) for s in d0]
    untagged = 1 - sum(s['share'] for s in d0)
    if untagged > 0.005:
        rows.append(('(untagged)', f'{100 * untagged:.1f}%', round(untagged * a['toffoli'])))
    return pd.DataFrame(rows, columns=['stage', 'share', 'Toffolis']).assign(circuit=name)
def find(folder, stem):
    return next(p for p in (Path(folder) / f'{stem}.kmb', Path(folder) / f'{stem}.kmx') if p.exists())
print('helpers ready')
```

```python
REF_COST = {'invsqrt': 2076774.233715282, 'dqi': 24439463583672.93, 'dfthcsos': 3355599588.3155365}  # the sample circuits, verified the same way
_VCACHE = {}
# circuits verified offline with these same verifiers (2026-10-26), keyed by the sha256 verify() computes below;
# a byte-identical circuit is not run again here (the slow part: invsqrt took 590 s and dqi 195 s on Kaggle's CPU)
VERIFIED = {
    'bdee18489fac2b38b86c2aab225b6469477c67a61ab8b2754f72c1f4a8f6ebe9': dict(score=385311255783.3984, num_qubits=892),     # dqi, SomeJoint's 23.77 (2026-11-01 06:05 run): their 892-qubit decoder with Ali Haydar Özdağ's flag carry on yoyobar's one-carry history, verified 2026-11-01 (logs/cron/2026-11-01-08_verify_dqi_somejoint-bdee1848.log)
    'fb2da4f299c4d41e421e5e57d07915599617291ab1c41d4e0fcefa6db1f2079c': dict(score=387059207747.3355, num_qubits=893),     # dqi, ahoirg's (Ali Haydar Özdağ) 893-qubit build of SomeJoint's decoder, shipped by ahoirg 23.64, heliosli (2026-11-01 05:40), manwithacat Real $150K and Leo Tang, verified 2026-11-01 (logs/cron/2026-11-01-08_verify_dqi_heliosli-fb2da4f2.log)
    '7dcb493e9de4bee7317b27772724ff778070f1d84d13408c853e2f0bab29dd56': dict(score=389453898406.98724, num_qubits=898),     # dqi, Ertuğrul Özer's TOP-100 23.650 (2026-11-01 02:23 run), a copy of SomeJoint's 898-qubit notebook; not a candidate row, verified 2026-11-01 (logs/cron/2026-11-01-08_verify_dqi_ozertuu-7dcb493e.log)
    '681a8a8da3131dcd2cef5bbf9fa6e3e4076d34e02448d6d44d37f29671535761': dict(score=327053613.2408534, num_qubits=428),     # dfthcsos, heliosli's 2026-11-01 05:40 run: Lera's (marculera) (9, 143, 40) factorization and angle circuits, their newer table weights, lambda 47.0619, 16,237 Toffolis, verified 2026-11-01 (logs/cron/2026-11-01-08_verify_chem_heliosli-681a8a8d.log)
    'f2088294dc71e3f88253a5be27bb634760d5d169dea6cbcece218f2a91f5190c': dict(score=327300810.34268564, num_qubits=428),     # dfthcsos, heliosli's version-33 weights on Lera's (9, 143, 40) factorization, as SomeJoint 23.77 ships them, lambda 47.0975, 16,237 Toffolis, verified 2026-11-01 (logs/cron/2026-11-01-08_verify_chem_somejoint-f2088294.log)
    '69da51111fd80329a1090aa32be4e87881eedd3cc1fd2a18ab59b837587d6fbc': dict(score=327359144.12415844, num_qubits=428),     # dfthcsos, manwithacat Real $150K's 2026-11-01 pick (also Leo Tang's), lambda 47.1059, 16,237 Toffolis, verified 2026-11-01 (logs/cron/2026-11-01-08_verify_chem_manwithacat-69da5111.log)
    '4e1d04245627ed1a70986af6ad038d84357b1a7344f0ab0063534fcec96c30a0': dict(score=328711157.2734612, num_qubits=428),     # dfthcsos, heliosli's 2026-10-31 23:40 run: Lera's (marculera) (9, 143, 40) factorization and angle circuits, their table weights, lambda 47.2074, 16,269 Toffolis, verified 2026-11-01 (logs/cron/2026-11-01-02_verify_chem_heliosli-4e1d0424.log)
    '76d06784447204713953319ef13b1aa390d4bba8dff6ad409f66d43bc5f8d374': dict(score=328917660.02085274, num_qubits=428),     # dfthcsos, Isabe11aM00re's QCC v49 (2026-10-31 21:17 run), lambda 47.2370, 16,269 Toffolis; dearer than heliosli's, not a candidate row, verified 2026-11-01 (logs/cron/2026-11-01-02_verify_chem_isabe11am00re49-76d06784.log)
    '9905991e7df07891cf1b2c0832dd70286a265c26260acc19e18d8d67f7a07e57': dict(score=330542464.24540395, num_qubits=428),     # dfthcsos, Isabe11aM00re's QCC v47 (2026-10-31 19:07 run), lambda 46.9766, 16,440 Toffolis; dearer than heliosli's, not a candidate row, verified 2026-11-01 (logs/cron/2026-11-01-02_verify_chem_isabe11am00re47-9905991e.log)
    '6b63ab57f1c60febb2a5cf48b037bc4525ac06a3887ad25ce549aa9abc5b9926': dict(score=392199879.9076627, num_qubits=423),     # dfthcsos, Soumitra Chavan's 18.75 notebook (2026-11-01 00:02 run), lambda 47.4993, 19,520 Toffolis; not a candidate row, verified 2026-11-01 (logs/cron/2026-11-01-02_verify_chem_soumitrachavan-6b63ab57.log)
    '909410f50a54b73844755fe1b32f898722ba036e1cad074b1f2003f2b60432f3': dict(score=393389075235.9089, num_qubits=900),     # dqi, Ahmed Berat Özer's v18: SomeJoint's 900-qubit decoder with his 576 history caches (2026-10-31 17:30 run), 900 qubits, 0 value / 0 phase errors, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_dqi_ahmedberatozer-909410f5.log)
    'de159897b507b16ec5e7591ef64ce7b590d7506d8f03e1a3c2dc0e23bbf28f5f': dict(score=331076721.7501548, num_qubits=428),     # dfthcsos, Ahmed Berat Özer's v18: his λ polish of Lera's C42 table, lambda 47.0525, 16,440 Toffolis, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_chem_ahmedberatozer-de159897.log)
    '38f0ac173b0f8e98620d9ffa4c44f3653da3b483d6d0610323b24cd11999b39d': dict(score=393944554828.0952, num_qubits=900),     # dqi, SomeJoint's (Agniv Banerjee team) 900-qubit decoder, narrow return (2026-10-31 16:19 run; also in Real $150K and Isabe11aM00re v47), 0/0 errors, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_dqi_somejoint-38f0ac17.log)
    'c64292ada641a5c69fbe04638e65b5cbd9299338cf9fdc4d62be4f6b550d4bb8': dict(score=331025488.42506784, num_qubits=428),     # dfthcsos, manwithacat's Hedge Wizardry HW-04: refit of Lera's (marculera) C42 weights, lambda 46.9767, 16,464 Toffolis, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_chem_manwithacat150k-c64292ad.log)
    'fe98677c61efd6f91ca7967254d8d415a23080326c1659c146bdb30c1915dedb': dict(score=398268953060.8457, num_qubits=888),     # dqi, manwithacat's Hedge Wizardry 16:29 run, 888 qubits, 0/0 errors, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_dqi_manwithacathw-fe98677c.log)
    '3d0aecffedb1f32179e8bb0126bcea622b5c7ee9169c68fbf660e49a0ad1ef7a': dict(score=402976561878.24097, num_qubits=891),     # dqi, Atakan Aldemir's 14:40 run, 891 qubits, 0/0 errors, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_dqi_atakanaldemir-3d0aecff.log)
    '7dd14481718dc9f34e84b7d74add1f7187922cd8fee46d09f4bfb1411985a604': dict(score=332967475.83147526, num_qubits=428),     # dfthcsos, Atakan Aldemir's 14:40 run, lambda 47.0579, 16,532 Toffolis, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_chem_atakanaldemir-7dd14481.log)
    'e574ad7512fee8b5c088d99d2f50aa4d71d5533cf68303d9a10d986922014d91': dict(score=333395100.914337, num_qubits=428),     # dfthcsos, Leo Tang's 17:21 run, lambda 47.1183, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_chem_leotang09-e574ad75.log)
    '32e30c08875959c2ad1c09cd706253a917d9def52cc9b795dd1b531ec68aa748': dict(score=398941122234.2813, num_qubits=888),     # dqi, Dmitrii Gluzdov's 15:48 run, 888 qubits, 0/0 errors, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_dqi_dmitriigluzdov-32e30c08.log)
    '3eca89fcdd08138ee50d2c577b6c5963997fafda2da4c3e73e08455d134b57de': dict(score=337714978.8792064, num_qubits=430),     # dfthcsos, Dmitrii Gluzdov's 15:48 run, lambda 48.2838, 430 qubits, verified 2026-10-31 (logs/cron/2026-10-31-20_verify_chem_dmitriigluzdov-3eca89fc.log)
    'c21f410afa7b0862366e9f7f57737b73427c554f46b38d1da058b3220728c6bb': dict(score=333376684.3481565, num_qubits=428),     # dfthcsos, Atakan Aldemir's C43 native compiler (2026-10-31 12:05 UTC run), 16,532 Toffolis, verified 2026-10-31 (logs/pub_261031e_verify.log)
    'b0f2c0e8467211b9d5a897a1830226c08d6a61cf43e9aafa1200156cd0cdce70': dict(score=403042819444.302, num_qubits=891),     # dqi, Atakan Aldemir's gap-3 residue repair (same run), verified 2026-10-31 (logs/pub_261031e_verify.log); dearer than yoyobar's e20e6abe
    'e20e6abea7e2167e27521929ff3101bf139b444bc94a67e56c1689830e70aff8': dict(score=398391039710.5484, num_qubits=888),     # dqi, yoyobar's 2026-10-31 11:52 run, 0 value / 0 phase errors, verified 2026-10-31 (logs/cron/2026-10-31-14_verify_dqi_yoyobar-e20e6abe.log)
    'bbbc7337f7fd6dce75cbdf8e2ba0a53d1d41daab22d9176d9ba3fca9f09cf6fd': dict(score=337521609.31504345, num_qubits=430),     # dfthcsos, heliosli's 2026-10-31 10:20 run, lambda 48.2561, 16,266 Toffolis, verified 2026-10-31 (logs/cron/2026-10-31-14_verify_chem_heliosli-bbbc7337.log)
    'e2d2b8b09087ab6938be7e8d79b34066edc20f7e589ae3de038881d1a7a88a8b': dict(score=337566911.14443934, num_qubits=430),     # dfthcsos, sigmaborov's Exact Gauge Search (2026-10-31 09:35 run), lambda 48.2626, verified 2026-10-31 (logs/cron/2026-10-31-14_verify_chem_sigmaborov-e2d2b8b0.log); dearer than heliosli's, not a candidate row
    '90f3e87790a865c2587bf44e644898d0a672f7e08240d9d8c87793e61e1db306': dict(score=337566987.22221684, num_qubits=430),     # dfthcsos, heliosli's 2026-10-31 08:08 run, lambda 48.2626, 16,266 Toffolis, verified 2026-10-31 (logs/cron/2026-10-31-09_verify_chem_heliosli-90f3e877.log)
    'c321189e8ef3fbc0615189bb94b2e7803bad36b5be99d5125005952a70abb7e3': dict(score=337780039.22259027, num_qubits=430),     # dfthcsos, sigmaborov's Exact Gauge Search (2026-10-31 08:44 run), verified 2026-10-31 (logs/cron/2026-10-31-09_verify_chem_sigmaborov-c321189e.log); dearer than heliosli's, not a candidate row
    'bce2948ff24e20fe59666d131f7fc4316a905103937272c52fdd39d494b6fe47': dict(score=100568.4991945597, num_qubits=62),     # invsqrt, Ehan Mallick's two 3-to-2 AND merges on yoyobar v8's 62-qubit circuit (2026-10-31 07:12 run), verified 2026-10-31 (logs/cron/2026-10-31-09_verify_inv_ehanmallick-bce2948f.log)
    '25db915d86ab9bf2cf41670ee97b7d32495fcb12f12aa48e89c1a5924e8921ec': dict(score=400273711353.48145, num_qubits=888),     # dqi, yoyobar's 888-qubit compiled candidate (2026-10-31 07:01 run, on SomeJoint's and heliosli's decoder line), verified 2026-10-31 (logs/cron/2026-10-31-09_verify_dqi_ehanmallick-25db915d.log)
    '43727d3f13774d495a9e13a116767aba4844e2700aff454e3f1b787c3af8d073': dict(score=337780213.3442293, num_qubits=430),     # dfthcsos, Sargis Khachatryan's Spectral Trust Epigraph (2026-10-31 04:14 run), shipped by yoyobar since 07:01, verified 2026-10-31 (logs/cron/2026-10-31-08_verify_chem_avazaku-43727d3f.log)
    'd4a8f109cc499879626b71ee241ed1883e039b031b65c8167b367dfbd23cb80d': dict(score=337651009.063524, num_qubits=430),     # dfthcsos, manwithacat's Hedge Wizardry HW-03 (2026-10-31 06:42 run): his constrained refit + column gauge of Özer's weights from Isabe11aM00re's frame, lambda 48.2746, 16,266 Toffolis, verified 2026-10-31 (logs/cron/2026-10-31-08_verify_chem_manwithacat-d4a8f109.log)
    '3821c71a905065148b3b28e3e6e6de85e2e8e6a7e72232996458ca79f7c02ae3': dict(score=337779790.59433043, num_qubits=430),     # dfthcsos, heliosli's 2026-10-31 06:17 run, lambda 48.2930, verified 2026-10-31 (logs/cron/2026-10-31-08_verify_chem_heliosli-3821c71a.log)
    'af180f734f787925800a500e558f54dc36f6533a6217130addcc922036f34ef8': dict(score=402702373901.5095, num_qubits=889),     # dqi, somejoint's (Agniv Banerjee team) 889-qubit decoder, shipped by heliosli and Hedge Wizardry since 2026-10-31, verified 2026-10-31 (logs/cron/2026-10-31-02_verify_dqi_somejoint-af180f73.log)
    '2366b67cf072a581776fe903591e9a02330d257cd26f00fb32517825291fdbd6': dict(score=100692.55756592617, num_qubits=62),     # invsqrt, yoyobar's candidate 1599 (dataset yoyobar/qcc-local-sota-20261031 v2) as shipped by Leo Tang's T7421 (2026-10-30 23:34 run), 62 qubits, verified 2026-10-31 (logs/cron/2026-10-31-02_verify_inv_leotang09-2366b67c.log)
    'b684dc6f228c29b3072c5a20252f4278dcd223aba62ffe6dc52ade9687850050': dict(score=337780864.9643602, num_qubits=430),     # dfthcsos, heliosli's 2026-10-31 00:05 run: their new table weights (lambda 48.293) on Atakan Aldemir's word table and the 16,266-Toffoli per-slot loaders, verified 2026-10-31 (logs/cron/2026-10-31-02_verify_chem_heliosli-b684dc6f.log)
    '5faba45918538eec20534fc143bb9b2ec5d91931a39038550e4cd37e4fd713a0': dict(score=401986129571.4192, num_qubits=888),     # dqi, manwithacat's VQCC Hedge Wizardry entry HW-01 (2026-10-30 22:06 run): yoyobar v8's 889-qubit decoder (73e1b187) with the epoch-0 pole qubit released and recomputed, 888 qubits, verified 2026-10-31 (logs/cron/2026-10-31-02_verify_dqi_heliosli-5faba459.log)
    'b52f5d6daef71a3120074374d4377164d8c358fa5a5d11202584485feb51682f': dict(score=337798090.4640843, num_qubits=430),     # dfthcsos, Özer's v16 (2026-10-30 15:37 run): his new weights (lambda 48.296) on Atakan Aldemir's word table and the 16,266-Toffoli loaders, verified 2026-10-30 (logs/cron/2026-10-30-20_verify_chem_manwithacat150k-b52f5d6d.log)
    '415268615c1f2e4f14c26e54e56cba2aa5c047a705d6e9561c1699d3f4134218': dict(score=337823023.2223082, num_qubits=430),     # dfthcsos, yoyobar's v8 (2026-10-30 18:10 run): new weights (lambda 48.299) on the 16,266-Toffoli loaders, verified 2026-10-30 (logs/cron/2026-10-30-20_verify_chem_yoyobar-41526861.log)
    '73e1b187b288f3cdc7d345214dd78ae6857d2f419041caf02bf3e29d26f7e6f2': dict(score=402458290261.8694, num_qubits=889),     # dqi, yoyobar's v8 (2026-10-30 18:10 run): his own 889-qubit builder on somejoint's decoder line (not spliced from manwithacat's 888-qubit file), verified 2026-10-30 (logs/cron/2026-10-30-20_verify_dqi_manwithacat150k-73e1b187.log)
    '67482db6c53896348eadba126924868ed3011d7e1576a6d0abb6f7c137ae4997': dict(score=100692.58221354822, num_qubits=62),     # invsqrt, yoyobar's v8 (2026-10-30 18:10 run): his own 100,754.54 candidate combined with Özer v15's phase patch, 62 qubits, verified 2026-10-30 (logs/cron/2026-10-30-20_verify_inv_manwithacat150k-67482db6.log)
    '84fcab7eb04eea0b5ad88184a3a427b72df73ea3499436c6d2d4df09dbd46d87': dict(score=337833376.1326832, num_qubits=430),     # dfthcsos, manwithacat's VQCC Quantum Voodoo v4 (2026-10-30 15:00 run): his lambda refit (48.301) of heliosli's weights on v60's 15 angle loaders, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-16_verify_chem_manwithacat-84fcab7e.log)
    '78d5786516b239514003ec62b037ae13707a006172c78927497a2e79933b7b48': dict(score=337991954.0482345, num_qubits=430),     # dfthcsos, heliosli's newest (2026-10-30 13:43 run): new weights (lambda 48.323) on v60's 15 angle loaders, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-16_verify_chem_heliosli-78d57865.log)
    '2ed7ccf189f13b71f6ff64ade4d6bbadd01bfe61e93c038a39158ccc83f29555': dict(score=403339761880.02136, num_qubits=888),     # dqi, manwithacat's VQCC Quantum Voodoo v4 (2026-10-30 15:00 run): his 888-qubit decoder on somejoint's and heliosli's 889-qubit line, groups 4-7 rebuilt as homogeneous chains (the failure fix), verified 2026-10-30 (logs/cron/2026-10-30-16_verify_dqi_manwithacat-2ed7ccf1.log)
    'dd111686c3e764cd77ead9e419840c8ea9714ff8ae125a9577ca3b552f2b1268': dict(score=337993966.8794745, num_qubits=430),     # dfthcsos, heliosli's newest (2026-10-30 08:43 run): new weights (lambda 48.324) on v59's 15 angle loaders, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-14_verify_chem_heliosli-dd111686.log)
    '315bf522a53c42681f4f9a4a34a93e9d12ff23426719fe5a1810e544eb8c2080': dict(score=338047050.9649815, num_qubits=430),     # dfthcsos, Ahmed Berat Özer's v15 (2026-10-30 11:12 run): lambda polish of v59's chemistry weights (lambda 48.331), 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-14_verify_chem_ahmedberatozer-315bf522.log)
    '6dbb98b34c3b16f1c4e019350c10d42259a6faa97e464c9b7c9390f877f392f4': dict(score=100940.56165415398, num_qubits=62),     # invsqrt, Ahmed Berat Özer's v15 (2026-10-30 11:12 run): yoyobar's 62-qubit circuit with one Toffoli removed by a history cache, verified 2026-10-30 (logs/cron/2026-10-30-14_verify_invsqrt_ahmedberatozer-6dbb98b3.log)
    '731944f2e94ddf2c22bcd0b49b9553bd3b7ae02445fa84e07345df6bbc9d9cf1': dict(score=338073281.9425884, num_qubits=430),     # dfthcsos, heliosli's newest (2026-10-30 04:36 run): new weights (lambda 48.335) on v58's 15 angle loaders (byte-identical), 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_heliosli-731944f2.log)
    'aa2c76f73dc94bb270b37272fb534decbef96b4e4c7d3e33f8526737edd75b1f': dict(score=338130690.99272305, num_qubits=430),     # dfthcsos, Pablo Diego Pereira's starter (2026-10-30 04:41 run): Gluzdov v44's chemistry re-gauged, lambda 48.343, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_pablodiegoo-aa2c76f7.log)
    '9c93ad9bda1d354a78f158eff81a8491f19da1a4cfbb8b729314e21dade2db50': dict(score=338130690.99867845, num_qubits=430),     # dfthcsos, Dmitrii Gluzdov's newest (2026-10-30 04:39 run, his v44; the same bytes in yoyobar's): lambda 48.343, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_yoyobar_gluzdov-9c93ad9b.log)
    'b08e0a5243f42d2107b2bc9dc5b15974a1989b49195822e5ef3528d92c869032': dict(score=338242144.68460536, num_qubits=430),     # dfthcsos, Dmitry Abulkhanov's cand-0044 (2026-10-30 04:39 run): lambda 48.359, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_dmitriyab0044-b08e0a52.log)
    'd74be3cb74b6c2ef43ad76adc15c77574b505cf3a2fd86756588a230695839bf': dict(score=338242226.8170634, num_qubits=430),     # dfthcsos, Sargis Khachatryan (avazaku)'s Epigraph Rank Scales (2026-10-30 03:50 run): lambda 48.359, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_avazaku_epigraph-d74be3cb.log)
    '2785399bfb8f9b8f2550bca0608050534882647bc321c5279996ae9ecc01d17c': dict(score=338242238.24466527, num_qubits=430),     # dfthcsos, Sargis Khachatryan (avazaku)'s Free-Gap SQP (2026-10-30 01:33 run): lambda 48.359, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_avazaku_freegap-2785399b.log)
    'f66332a9a0fcaa6e4b31ba914633f14a79a5944f24c8a02f8bfb1c5a9df990b2': dict(score=338938769.01246125, num_qubits=430),     # dfthcsos, Dmitry Abulkhanov's cand-0040 (2026-10-30 00:59 run): lambda 48.459, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_dmitriyab0040-f66332a9.log)
    '1cb80270bd70b66f1d44246ddf455c1156b7e2dc97ac214c86e514fab20e93d0': dict(score=3192441045.3012543, num_qubits=742),     # dfthcsos, Leo Tang's T7421 (2026-10-30 02:14 run): lambda 49.301, 87,269 Toffolis, 742 qubits, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_chem_leotang-1cb80270.log)
    'a568bc79348165bad5be2349960d200300c5639ffd2142aca6d548e764a399d9': dict(score=101002.47843760978, num_qubits=62),     # invsqrt, yoyobar's 62-qubit circuit (busyaprime's 63-qubit line narrowed by one qubit), the same bytes in heliosli's, Gluzdov's and Pablo Diego Pereira's newest, verified 2026-10-30 (logs/cron/2026-10-30-08_verify_invsqrt_yoyobar-a568bc79.log)
    'c57ba2328586f2a5d3b4c439c6a8970c6477b401b411d16ea0b4ef7d5f5bd839': dict(score=475538.71152276226, num_qubits=145),     # invsqrt, Leo Tang's T7421 (2026-10-30 02:14 run), verified 2026-10-30 (logs/cron/2026-10-30-08_verify_invsqrt_leotang-c57ba232.log)
    '20b1e677f5af91daec235a9866c237084a441284022719301ed4fdb68663743f': dict(score=338248976.51953065, num_qubits=430),     # dfthcsos, heliosli's newest (2026-10-29 23:39 run): new weights (lambda 48.360) on the 15 angle loaders of manwithacat's v57 per-slot assembly (byte-identical), 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-02_verify_chem_heliosli-20b1e677.log)
    '74566610c218752a428e9048029fc0d7860ec54c5349bc6c08fc0735b83365cd': dict(score=338879606.6099476, num_qubits=430),     # dfthcsos, Dmitrii Gluzdov's newest (2026-10-30 00:02 run): lambda 48.450, 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-02_verify_chem_gluzdov-74566610.log)
    '7c6d1ad0a1d72757cd6835b4ef54db1036630c795e66900223bfbdf359f8bb92': dict(score=338938769.01246125, num_qubits=430),     # dfthcsos, manwithacat's newest (2026-10-29 21:19 run): lambda 48.459 on Gluzdov's newest loaders (byte-identical), 16,266 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-02_verify_chem_manwithacat-7c6d1ad0.log)
    'cd75f18f6947cd33b83269191d5edae60490a0a6f54151220b3da8cb9ac471db': dict(score=340231509.3341443, num_qubits=430),     # dfthcsos, Sargis Khachatryan (avazaku)'s Active-Face SQP (2026-10-29 20:30 run): lambda 48.614, 16,276 Toffolis, verified 2026-10-30 (logs/cron/2026-10-30-02_verify_chem_avazaku-cd75f18f.log)
    '363a2d0e5cffeac441af1858242830992db73b5b3ddef4cbfa92c4d1a3f66c8a': dict(score=339059918.2297551, num_qubits=430),     # dfthcsos, manwithacat's per-slot assembly (2026-10-29 18:37 run): busyaprime's newest weights (lambda 48.476, submission.npz byte-identical) with the cheapest public loader per slot (2, 3, 14 Gluzdov's newest; 0, 1, 4, 5, 8 heliosli 22.81's; the rest Atakan Aldemir's V3), 16,266 Toffolis, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_manwithacat-363a2d0e.log)
    'e31463ffc98d9a3d439320f0d900b34c17adf2ed6959b04f6501869b03de95a8': dict(score=339143297.0366479, num_qubits=430),     # dfthcsos, busyaprime 22.87's (2026-10-29 17:54 run): his weights (lambda 48.476) on Atakan Aldemir's V3 loaders, 16,270 Toffolis, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_busyaprime-e31463ff.log)
    '947a73f68d038d89c6242e912bc519e481b73412debdf77aa0800520ba51e65a': dict(score=339145888.7809303, num_qubits=430),     # dfthcsos, Ahmed Berat Özer's newest (2026-10-29 16:06 run): his weights (lambda 48.476) on Atakan Aldemir's V3 loaders, 16,270 Toffolis, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_ozer-947a73f6.log)
    '1544b143c44cd96b6084950153b94e92311ceb4e99259976b815115fefe89b9f': dict(score=339519495.5323955, num_qubits=430),     # dfthcsos, sigmaborov 22.86's exact gauge search (2026-10-29 17:36 run): lambda 48.530 on Atakan Aldemir's V3 loaders, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_sigmaborov-1544b143.log)
    'b539eedf5c1552c407757a4186ad829f95702ace72c903e5c88541172c8ef3c6': dict(score=339569433.02475107, num_qubits=430),     # dfthcsos, Dmitrii Gluzdov's newest (2026-10-29 16:31 run): lambda 48.549, 16,266 Toffolis, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_gluzdov-b539eedf.log)
    'cbcd641b34adfced8c353b1be99d10fc202e571db22bec3759ed0ae10311d588': dict(score=339704716.6764897, num_qubits=430),     # dfthcsos, Baidalin Adilzhan's (2026-10-29 16:26 run): lambda 48.556 on Atakan Aldemir's V3 loaders, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_baidalin-cbcd641b.log)
    '2e0fcafe8e421ea54fe2446ba7b8ff6b65735985e7fc2a0759918a454ea9a8fc': dict(score=339861567.46676314, num_qubits=430),     # dfthcsos, Nilanjan Saha's Auto-Assemble (2026-10-29 14:27 run): lambda 48.573 on heliosli 22.81's loaders, verified 2026-10-29 (logs/cron/2026-10-29-20_verify_chem_nilanjan-2e0fcafe.log)
    'd04595e48cc2cc1fe7aca7bfbdee1997598b5c6a36ecba47d5997b9f0a7683df': dict(score=339728082.3378603, num_qubits=430),     # dfthcsos, Atakan Aldemir's source-built chemistry V3 (2026-10-29 13:35 run): learned factor (lambda 48.560) and 15 loaders at 16,270 Toffolis, with heliosli's two and Gluzdov's one AND reductions; kirilltym and manwithacat ship the same bytes, verified 2026-10-29 (logs/cron/2026-10-29-16_verify_chem_d04595e4.log)
    '7a378f4755605e0c5fa6117f808891bf8dcf0797d047155e6d4c9e75e3c99421': dict(score=410274633000.4907, num_qubits=889),     # dqi, hengck23's 889-qubit approximate decoder (dqi_iter014, projective extraction; the author discloses rare-weight failures), shipped by kirilltym's Exact Gauge + 5-Word QROM (2026-10-29 14:08) and manwithacat's Real $150K (14:47), verified 2026-10-29 (logs/cron/2026-10-29-16_verify_dqi_kirilltym-7a378f47.log)
    '736f3055850794c6beaedf8de5f7dafa1c270e2b57edd403ffa96abf6d207741': dict(score=412372251652.37823, num_qubits=888),     # dqi, Atakan Aldemir's 888-qubit node-policy repair of somejoint 22.77's 889-qubit decoder (2026-10-29 13:35 run), verified 2026-10-29 (logs/cron/2026-10-29-16_verify_dqi_atakan-736f3055.log)
    '826ee9d59345b42835b0292115b8ee860ff6e4ab53c564f4c37f2b1c608a8d7e': dict(score=339861567.46676314, num_qubits=430),     # dfthcsos, heliosli 22.81's 15 angle circuits (16,272 Toffolis) with Gluzdov 22.81's submission.npz (lambda 48.573), built in section 1, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_chem_combo-helio-gluzdov.log)
    '18ca296299b3cef2a5f364a2a12bbc5460798c5a6516264df1d435068df2358c': dict(score=339903340.0291362, num_qubits=430),     # dfthcsos, Gluzdov 22.81 (2026-10-29 12:58 run): his own angle circuits (16,274 Toffolis) and a new gauge, lambda 48.573, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_chem_gluzdov-18ca2962.log)
    '73b81dda389144aef0fe5ffd737360bb5db97f7feb603af24c4d6e1a009091bb': dict(score=339980573.1697358, num_qubits=430),     # dfthcsos, leoprovorov Full-XOR Chemistry (2026-10-29 12:54 run): heliosli 22.81's circuits with his W/H2_I calibration, lambda 48.590, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_chem_leoprovorov-73b81dda.log)
    '72306a5efe1ad6ae766a0d944b59b86f4f57455003526388abee6050879625c1': dict(score=339980584.97930497, num_qubits=430),     # dfthcsos, heliosli 22.81 (2026-10-29 10:57 run): their newest 15 angle circuits, 16,272 Toffolis, lambda 48.590, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_chem_heliosli-72306a5e.log)
    '0bce84510f4ab9a915bae3c664e17701283cb36690ca98a75d18910a4cbce425': dict(score=340064159.36105996, num_qubits=430),     # dfthcsos, Özer's group compression (2026-10-29 12:06 run): 16,276 Toffolis, lambda 48.590, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_chem_ozer-0bce8451.log)
    'b143fe4b9847599f6b511f0bed6085aaa25a775ce4e4f619ebde0680a4aff014': dict(score=340232090.43014985, num_qubits=430),     # dfthcsos, somejoint 22.77 (2026-10-29 10:35 run): their factorization with Gluzdov's table weights, as heliosli shipped it, lambda 48.614, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_chem_somejoint-b143fe4b.log)
    '6349d2bc82eb1ef3a4e8c52c71a312823d756532517463b5e18eb25cc81195a8': dict(score=414211737069.8777, num_qubits=889),     # dqi, somejoint 22.77's 889-qubit decoder (their team's third line, 2026-10-29 10:35 run), 1 phase error at the verifier's defaults here (somejoint's log: the same 4.14212e11); Özer's group compression (12:06) and Gluzdov 22.81 (12:58) ship the same bytes and log 4.12318e11 with 0 errors; dearer here than Chaturvedi's, verified 2026-10-29 (logs/cron/2026-10-29-14_verify_dqi_ozer-6349d2bc.log)
    'b5d50e8948d3739295285e750a1ef53073f9d6283c78f8be3dba1a167a28e61c': dict(score=340441129.56165034, num_qubits=430),     # dfthcsos, heliosli 22.75's 15 angle circuits (16,286 Toffolis) with Gluzdov 22.75's submission.npz (lambda 48.614), built in section 1; atakanaldemir's Iso-Gap (06:47 run) ships the same bytes, verified 2026-10-29 (logs/cron/2026-10-29-08_verify_chem_atakan-b5d50e89.log)
    '952da52cdf2d21ee578a7ee7922352a8c2fe5661e868672a15c47f7b7915d687': dict(score=340571920.8004815, num_qubits=430),     # dfthcsos, heliosli 22.75 (2026-10-29 03:27 run): their newest 15 angle circuits, 16,286 Toffolis, lambda 48.632, verified 2026-10-29 (logs/cron/2026-10-29-08_verify_chem_heliosli-952da52c.log)
    'f8a62242dd671ef061e18220e5ab3a8734bcc2010cb0f976202b5c32b9e186f2': dict(score=340650168.69315076, num_qubits=430),     # dfthcsos, Gluzdov 22.75 (2026-10-29 05:16 run): his own angle circuits (16,296 Toffolis) and a new gauge, lambda 48.614, verified 2026-10-29 (logs/cron/2026-10-29-08_verify_chem_gluzdov-f8a62242.log)
    'b2f5c70547188bed604ebbc288d799f9976d2c4ba0de838ccd355be8123c6da0': dict(score=341342227.7452543, num_qubits=430),     # dfthcsos, avazaku Slack Homotopy (2026-10-29 02:08 run): heliosli v12 circuits, his gauge, verified 2026-10-29 (logs/cron/2026-10-29-08_verify_chem_avazaku-b2f5c705.log)
    'de2d17dc7760974779178af6d96f0ac2c97babdf887011b33ca8b729182a05e7': dict(score=412645695849.98254, num_qubits=895),     # dqi, Mudit Chaturvedi's 895-qubit streamed decoder (somejoint 22.75's line, node range 10..16), shipped in manwithacat's Real $150K (2026-10-29 04:57 run), verified 2026-10-29 (logs/cron/2026-10-29-08_verify_dqi_muchat-de2d17dc.log)
    '9a8627c812833a6feca6eb82239f38d1a9a5e1eee299a9a7af9abd85e36339be': dict(score=102349.2360480486, num_qubits=63),     # invsqrt, busyaprime 22.72 (his new 63-qubit circuit), verified 2026-10-29 (logs/cron/2026-10-29-02_verify_invsqrt_busyaprime-22-72431.log)
    'da17bb3859bf11ffa05f73d7d68849f8327ec314c726666c32fd749eb14d078e': dict(score=340781040.24098283, num_qubits=430),     # dfthcsos, heliosli 22.74's 15 angle circuits (16,296 Toffolis) with Gluzdov 22.72's submission.npz (lambda 48.632), built in section 1, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_combo_helio_gluzdov.log)
    'fb68d058e776f6f65465cc0bad380e2df3cd5673cc105b0f81d52eed976a067c': dict(score=340990159.68148416, num_qubits=430),     # dfthcsos, Gluzdov from source (2026-10-29 00:04 run): the heliosli v12 / somejoint C60 line, 16,306 Toffolis, his W and H2_I, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_quantum-circuit-compilation-from-source.log)
    'e198421372a328a458a06808be2662738acd57f0175bc37999f5d21f13b38f49': dict(score=341135130.39491904, num_qubits=430),     # dfthcsos, atakanaldemir Iso-Gap: 16,296 Toffolis, U changed, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_quantum-circuits-iso-gap-dqi-h5.log)
    '33b03d9ae7951c38b7b91da0a75d34f8c85b3bb2f71e2a6b55abc6cda1357c17': dict(score=341375039.5769367, num_qubits=430),     # dfthcsos, Nilanjan 21.87+ (2026-10-28 22:25 run): heliosli v12 circuits, his gauge, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_21-87-fused-dqi-c46-chemistry-gold.log)
    '07a0e35265b11bc80f109e625cd0702262441e2b8f2116e5efc07e7a7e8d0369': dict(score=341375166.66254705, num_qubits=430),     # dfthcsos, somejoint 22.75: heliosli v12's circuits with kirilltym's exact gauge, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_quantum-circuits-22-75-local-gm-897-qubit-dqi.log)
    '24010424a1118cc3723fb51636290cbfb077c6c8e7755525f4cb310df49233ac': dict(score=341384301.8473315, num_qubits=430),     # dfthcsos, Vyas Senthilkumar Verified Assembly: heliosli v12 circuits with a gauge, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_verified-assembly-of-public-qcc-circuits.log)
    '51398fb0ef8c68751db95f32e5c035f730590b1010741ab123529b354ec9a8d3': dict(score=341880855.4426008, num_qubits=430),     # dfthcsos, busyaprime 22.72: heliosli v12 circuits, his gauge, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_chem_lb-22-72431-22x-cheaper-than-google-s-circuits.log)
    '53d8b2edf251e2a0c54c401a97d1396c7ed309e1cbfaab8d54a2c99a11ed6552': dict(score=412943252615.63153, num_qubits=897),     # dqi, somejoint 22.75: their team's 897-qubit decoder with a streamed conversion, verified 2026-10-29 (logs/cron/2026-10-29-02_verify_dqi_quantum-circuits-22-75-local-gm-897-qubit-dqi.log)
    '3243168661a40715710d2036e821362680025c10c25874943c055b41750fbf7e': dict(score=343172572.78225887, num_qubits=430),     # dfthcsos, somejoint 22.67: their team's new C60 factorization, full-XOR loaders, 16,366 Toffolis, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem2_quantum-circuits-22-67-local-gm-full-xor-chem.log)
    'aa8d643218376b0db28d3bae820c242f8067618f4a74c49d8553817d62f6a35f': dict(score=344957009.4068184, num_qubits=430),     # dfthcsos, avazaku C60 Gauge Descent: somejoint 22.63's circuits with avazaku's gauge, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem2_c60-gauge-descent-full-xor-qrom.log)
    'c7bdd23cf60cb2ac75b65012fcd07d694047ab59fcdfa712c7fb5fab3b6e376d': dict(score=343477342.01019543, num_qubits=430),     # dfthcsos, atakanaldemir 22.66's circuits with kirilltym 22.64's submission.npz, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_combo_atakan_kirill.log)
    'a4d7ec844f32af7baf2a3e382b41bcf63b7c58266d087785b6d37f46dc27b3af': dict(score=343787901.22637963, num_qubits=430),     # dfthcsos, atakanaldemir 22.66: somejoint 22.63's factorization, cheaper angle loaders, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_quantum-circuits-c60-reachable-cache.log)
    '3b0c29db29b1b7a8105725e63c5fe65aa0cae7efa520f0243bb88246c46d3dbf': dict(score=344812530.9004197, num_qubits=430),     # dfthcsos, kirilltym 22.64: somejoint 22.63's circuits with his exact gauge, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_quantum-circuits-exact-gauge-5-word-qrom.log)
    '3ad07c64288db5f571f34a01767a0c1d8132b3fdc5c11dbade7fd799beac5fb0': dict(score=345124297.3438777, num_qubits=430),     # dfthcsos, somejoint 22.63: their team's R8 B143 C60 factorization, full-XOR five-word lookups, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_quantum-circuits-22-63-local-gm-full-xor-chem.log)
    '5a4504761842b659d65759b590541e6c09761c4aa97072dcbab7f5f48b54b9a7': dict(score=345124286.35393983, num_qubits=430),     # dfthcsos, leoprovorov Full-XOR Chemistry (somejoint 22.63's line), verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_full-xor-chemistry-no-borrowed-bits.log)
    '129edc4f507165ba0f4d3844b45675f774a99eb58769f9fe4d21ef6464b9ce31': dict(score=347351539.41396075, num_qubits=430),     # dfthcsos, avazaku R8 Block Gauge, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_google-quantum-ai-r8-block-gauge-5-word-qrom.log)
    '2d65f4a3f00929e7a41d0331e999d227da7e87e8d38e845ef9818ac8e14225ba': dict(score=347361565.95524335, num_qubits=430),     # dfthcsos, heliosli 2026-10-28 09:11 run (also abisheksrivastav's audit), verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_quantum-circuit-portfolio-reproducibility-audit.log)
    '86c053a308b4814bc5688d9580433cfa6085edf84ce0a6ff6b23604b6aedd83a': dict(score=347715311.5963253, num_qubits=430),     # dfthcsos, leoprovorov Unbearable Lightness, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_the-unbearable-lightness-of-quantum-circuits.log)
    '21881dca9461f4c6383ea6ef2171375f076cf0d97b6c46cfce23d6d2c3e7d4a0': dict(score=347715359.9909825, num_qubits=430),     # dfthcsos, leoprovorov Width Lab, verified 2026-10-28 (logs/cron/2026-10-28-14_verify_chem_but-some-q-squared-width-lab.log)
    '4ee7a17b4cbca4053a19bc951e7f2ef91fabdbb036006c9d658adcff4fe76da4': dict(score=414781342770.7355, num_qubits=889),     # dqi, heliosli 22.54 (2026-10-28 04:28 run; also in Nilanjan 22.51), verified 2026-10-28 (logs/cron/2026-10-28-08_verify_dqi_heliosli.log)
    '992692a9e3f87869ba4118cba79b7ba3f8096dc4c589eafb0896627fccee6bdf': dict(score=347705355.9537195, num_qubits=430),     # dfthcsos, Nilanjan 22.51: somejoint 22.44's circuits with his column gauge, verified 2026-10-28 (logs/cron/2026-10-28-08_verify_chem_nilanjan.log)
    'bdaae31d3f1957733e6515c66d88bbd44439b8874af25a82001ff2c04479e571': dict(score=347715359.9909825, num_qubits=430),     # dfthcsos, somejoint 22.44: their team's R8 B143 factorization, phased five-word lookups, verified 2026-10-28 (logs/cron/2026-10-28-08_verify_chem_somejoint.log)
    'b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6': dict(score=422346240055.8793, num_qubits=897),     # dqi, somejoint 22.34: their 897-qubit decoder with the 658-Toffoli jet, verified 2026-10-28 (logs/cron/2026-10-28-02_verify_dqi.log)
    '07314803a0925970cf45007ebbdb8a146378c242e3991c70d6f7509af3d583da': dict(score=102664.23957721348, num_qubits=63),     # invsqrt, busyaprime 22.12 (63 qubits, from his 33-qcc-v8-members dataset), verified 2026-10-28 (logs/cron/2026-10-28-02_verify_inv.log)
    'ef334fee63f3367fe721549dd94d4264793c56e3c49a09177c05b353ac71c927': dict(score=436484409263.83356, num_qubits=883),     # dqi, somejoint 21.98: their 883-qubit decoder with a 658-Toffoli jet (also in heliosli 22.13), verified 2026-10-28 (logs/cron/2026-10-28-02_verify.log)
    '49aa3cdc2e9ba26e17b3beea551a8d0046e9a600bfef30e222d22c603fbec032': dict(score=349339717.3328993, num_qubits=431),     # dfthcsos, heliosli 22.13 (2026-10-28 00:11), verified 2026-10-28 (logs/cron/2026-10-28-02_verify.log)
    'f3d335c3f693aaab93bcea8750605f206dd2ba6d04ad5bf9a6de1de1ac5ec3fb': dict(score=444151956391.4917, num_qubits=877),     # dqi, Ahmed Berat Özer v9: somejoint's fused decoder narrowed to 877 qubits, verified 2026-10-27 (logs/verify_ozer_1027e.log)
    '7beea291a5de36ef2b4fb50585b2f8d7cce02d3c575b491cf5dcc93e9d9655f6': dict(score=448601212111.76025, num_qubits=889),     # dqi, somejoint 21.34 fused decoder (zero-power padding, retained norms), shipped byte-identical in heliosli J5, verified 2026-10-27
    '8e89eabc9c65a14234e54320970611aa1b806fbc6a4bec7d4483ea39fb128cad': dict(score=376433626.307489, num_qubits=431),     # dfthcsos, heliosli J5 newest: five-word circuits with deeper lookup plans on mrmeows's C59 table, verified 2026-10-27
    '39226d8f12b30348a586f67596134eea6ce8804937296cdc2d4e60b96c5b3730': dict(score=450382244807.3486, num_qubits=889),     # dqi, somejoint 21.20 confluent decoder with Taylor completion, verified 2026-10-27
    'd2bf396a87f3d91467aea22587584e56eddab71051228ce0abacb9561956d327': dict(score=383267019.2215811, num_qubits=431),     # dfthcsos, heliosli's five-word angle circuits on mrmeows's R10 table, verified 2026-10-27
    '0766a111a76f241aa57fbbed466436d743be13436943ddb0628c291ca07d67d6': dict(score=103546.168164559, num_qubits=63),     # invsqrt, mrmeows 21.0 (63-qubit hybrid, 2 Toffolis fewer), verified 2026-10-27
    '0048fb3263b8a0dd692df3feaf3f1174a5839bc0ec4658e4a19e892b4b7bb776': dict(score=389636552.06586975, num_qubits=423),     # dfthcsos, mrmeows 21.0 (column merge C 60 -> 59), verified 2026-10-27
    '3ed5f7fb321dec6a174c2e736dbe64b436b798f3021f02cacaa7961f636e090f': dict(score=103672.22039048169, num_qubits=63),     # invsqrt, mrmeows 20.39 (63-qubit hybrid), verified 2026-10-27
    '86a2aef42f7eaf98236aea08bcb933c819846308ce9367a82e7c1f5807c4fc90': dict(score=391389998.0646679, num_qubits=423),     # dfthcsos, mrmeows 20.39 (rustad's R10 factorization, shared-product lookups), verified 2026-10-27
    '0161fd2a6e50979838f8c4896b490ddea4d7a673588a9761a63a561a59b7ece5': dict(score=455954457807.33057, num_qubits=889),     # dqi, somejoint 20.96 confluent Hermite decoder, verified 2026-10-27
    '9ac50386192a544f353ab0f46604e5e072981b8bfc8521e8551c689a9445e3b9': dict(score=108560.35824641786, num_qubits=63),     # invsqrt, mrmeows open circuits (63 qubits)
    '4ba42bdfb94bb25434cbed0b84d1400f8b76079eb05e28897ffa7e2aefed6681': dict(score=501480157.9635119, num_qubits=440),     # dfthcsos, mrmeows open circuits (440 qubits)
    '545189b999f49a4d226480df711fe66dbf996250a92dcc3fc42c4bad6ca65822': dict(score=485193743459.73724, num_qubits=882),     # dqi, Özer v8's prefix and middle block with our 882-qubit reverse synthesis
    '569863ae0fafd5f94fe91b498ec67ed0a4d96b826ced31cb0ff57c7fe6c063cb': dict(score=495048250576.62756, num_qubits=883),     # dqi, Özer v8: our 883-qubit Schur build with mrmeows's round17 Bézout middle block
    '26369ae8dfaa62dd5963e4ce16c5bb6b3ba6dd1fa090ee322810a0bb78803d3e': dict(score=500874373991.25, num_qubits=883),     # dqi, Schur CF, 883-qubit reverse synthesis
    'd48fc426e636d549e7d1e52daef03bb1bf3e24b6ffc2da1385c26f11d8f8beeb': dict(score=515160355602.46826, num_qubits=892),     # dqi, Schur CF rebuilt from jday96314
    '908ba689a0b5b167c8fbe91c7d13cbb25075936975dff117861b1b78b536edf9': dict(score=114323.93229681085, num_qubits=66),     # invsqrt, mrmeows
    'e8083195cec294f2adde9f1b991bcc348233f549a49e5083ac25a12d0777385b': dict(score=842819878125.7235, num_qubits=880),     # dqi, mrmeows
    '562d1dd806b256c6bce88a365fdaee7c4e0e20207c27298adac79daf95d5d4d6': dict(score=815064701829.9167, num_qubits=880),     # dqi, Özer 2026-10-26 afternoon
    '93c18c155eef38474ff2f1cc029ac6a109ed4aa895d73041f5e8a73848ae9c05': dict(score=508908870.7302555, num_qubits=448),     # dfthcsos, Özer 2026-10-26 afternoon
    'd3d57207f6258936be8ea3f2221b555f90ed4066fef11cdeb76adc0d4cfb14e6': dict(score=425969844.4690444, num_qubits=423),     # dfthcsos, ours: rustad's factorization + mrmeows's compiler, J 4 under the 128-qubit cap
    '21bfa94414d1434b67f2b987d6850dcd0ace92181d4e11107c71c1ab955809e0': dict(score=436969552.20496726, num_qubits=415),     # dfthcsos, ours: rustad's factorization + mrmeows's angle compiler, J 3
    '5f83eb400d23a071ac7063c711b189e3309490cbfb8cdfda656e0a878a414d92': dict(score=494962019.3347939, num_qubits=440),     # dfthcsos, Özer 2026-10-26 evening (440 qubits, lambda 47.48)
    '719b6a0037956fd096c4c2f4d6cfc2deb8294d77a85c271e0e9eb5d3dc1462e5': dict(score=513261654.6882092, num_qubits=448),     # dfthcsos, Gluzdov
}
def verify(task, folder):
    t = time.time()
    f = Path(folder) / 'dfthcsos' if task == 'dfthcsos' else find(folder, task)
    h = hashlib.sha256(b''.join(p.read_bytes() for p in sorted(f.iterdir())) if f.is_dir() else f.read_bytes()).hexdigest()
    if (task, h) in _VCACHE:                                 # identical circuit already verified (portfolios share files)
        return dict(_VCACHE[task, h], sec=0)
    if h in VERIFIED:                                        # byte-identical to a circuit the same verifier passed offline
        return dict(VERIFIED[h], status='PASS', sec=0, offline=True)
    cmd = {'invsqrt': lambda: [VER / 'invsqrt' / 'verify_submission_invsqrt.py', find(folder, 'invsqrt')],
           'dqi': lambda: [VER / 'dqi' / 'verify_submission_dqi.py', find(folder, 'dqi')],
           'dfthcsos': lambda: [VER / 'dfthcsos' / 'verify.py', Path(folder) / 'dfthcsos']}[task]()   # only this task's file
    r = subprocess.run([sys.executable, '-u'] + [str(c) for c in cmd], capture_output=True, text=True, env=ENV)
    j = json.loads(r.stdout[r.stdout.index('{'):]) if '{' in r.stdout else {'status': 'NO_OUTPUT', 'err': r.stderr[-300:]}
    j['sec'] = round(time.time() - t); _VCACHE[task, h] = j; return j
# Every candidate was verified offline with these same verifiers (2026-10-26, one laptop); cost per task:
OFFLINE = pd.DataFrame({'mrmeows 20.39': [103672.22039048169, 495048250576.62756, 391389998.0646679],   # verified 2026-10-27; its dqi is Özer v8, byte-identical
                        'somejoint 20.96': [103672.22039048169, 455954457807.33057, 391389998.0646679],   # verified 2026-10-27; its invsqrt and chemistry are mrmeows 20.39's, byte-identical
                        'mrmeows 21.0': [103546.168164559, 455954457807.33057, 389636552.06586975],   # verified 2026-10-27; its dqi is somejoint 20.96's, byte-identical (listed after it, so a tie goes to its author)
                        'somejoint 21.98': [103546.168164559, 436484409263.83356, 355065066.0799456],   # verified 2026-10-28 (logs/cron/2026-10-28-02_verify.log); dqi theirs; invsqrt mrmeows 21.0's and chemistry busyaprime 21.88's, byte-identical
                        'heliosli 22.13': [103546.168164559, 436484409263.83356, 349339717.3328993],   # verified 2026-10-28; chemistry heliosli's; dqi somejoint 21.98's and invsqrt mrmeows 21.0's, byte-identical (listed after them, so a tie goes to the author)
                        'busyaprime 22.12': [102664.23957721348, 436484409263.83356, float('inf')],   # verified 2026-10-28 (logs/cron/2026-10-28-02_verify_inv.log): invsqrt his; dqi somejoint 21.98's, byte-identical; chemistry (his log: 3.5285e8) dearer than heliosli's, not verified here
                        'heliosli 22.54': [102664.23957721348, 414781342770.7355, 349339717.3328993],   # verified 2026-10-28 08 (logs/cron/2026-10-28-08_verify_dqi_heliosli.log); dqi theirs; invsqrt busyaprime's and chemistry their 22.13's, byte-identical
                        'somejoint 22.44': [102664.23957721348, 422346240055.8793, 347715359.9909825],   # verified 2026-10-28 08 (logs/cron/2026-10-28-08_verify_chem_somejoint.log); chemistry and dqi theirs; invsqrt busyaprime's, byte-identical
                        'Nilanjan 22.51': [103546.168164559, 414781342770.7355, 347705355.9537195],   # verified 2026-10-28 08 (logs/cron/2026-10-28-08_verify_chem_nilanjan.log); chemistry somejoint 22.44's with his gauge; dqi heliosli 22.54's and invsqrt mrmeows 21.0's, byte-identical (listed after them, so a tie goes to the author)
                        'busyaprime 22.72': [102349.2360480486, 410274633000.4907, 339143297.0366479],   # verified 2026-10-29 20 (logs/cron/2026-10-29-20_verify_chem_busyaprime-e31463ff.log); his 17:54 output (renamed LB 22.86841): invsqrt his (unchanged, logs/cron/2026-10-29-02_verify_invsqrt_busyaprime-22-72431.log), chemistry his new weights on Atakan's V3 loaders, dqi hengck23's (byte-identical); until v56: [102349.2360480486, 414781342770.7355, 341880855.4426008] (his 00:36 output)
                        'yoyobar 22.99': [100568.4991945597, 398391039710.5484, 337521609.31504345],   # verified 2026-10-31 14 (logs/cron/2026-10-31-14_verify_dqi_yoyobar-e20e6abe.log, logs/cron/2026-10-31-14_verify_chem_heliosli-bbbc7337.log); his 11:52 output (see EXTRA): dqi his, invsqrt Ehan Mallick's and chemistry heliosli's 10:20, byte-identical; until v67: [100692.58221354822, 400273711353.48145, 337780213.3442293]; verified 2026-10-31 09 (logs/cron/2026-10-31-09_verify_dqi_ehanmallick-25db915d.log, logs/cron/2026-10-31-08_verify_chem_avazaku-43727d3f.log); his 07:01 output (see EXTRA); until v65: [100692.58221354822, 402458290261.8694, 337823023.2223082]; verified 2026-10-30 20 (logs/cron/2026-10-30-20_verify_inv_manwithacat150k-67482db6.log, logs/cron/2026-10-30-20_verify_dqi_manwithacat150k-73e1b187.log, logs/cron/2026-10-30-20_verify_chem_yoyobar-41526861.log); his v8 (18:10 run); until v61: [101002.47843760978, 410274633000.4907, 338130690.99867845] (his 04:41 output)
                        'atakanaldemir 23.27': [100568.4991945597, 402976561878.24097, 332967475.83147526],   # verified 2026-10-31 20 (logs/cron/2026-10-31-20_verify_dqi_atakanaldemir-3d0aecff.log, logs/cron/2026-10-31-20_verify_chem_atakanaldemir-7dd14481.log); his 2026-10-31 14:40 run (Kaggle now mounts this one, not the 12:05 c21f410a): chemistry 7dd14481, dqi 3d0aecff, invsqrt Ehan Mallick's; until v70: [100568.4991945597, 403042819444.302, 333376684.3481565]; verified 2026-10-31 interactive (logs/pub_261031e_verify.log); his 12:05 UTC output (see EXTRA): chemistry and dqi his, invsqrt Ehan Mallick's, byte-identical (listed after yoyobar 22.99, so the invsqrt tie keeps yoyobar's row)
                        'Ehan Mallick 20.65': [100568.4991945597, 400273711353.48145, 337780213.3442293],   # verified 2026-10-31 09 (logs/cron/2026-10-31-09_verify_inv_ehanmallick-bce2948f.log); invsqrt his; dqi and chemistry yoyobar's 07:01 output, byte-identical (listed after it, so a tie goes to yoyobar)
                        'somejoint 22.75': [102664.23957721348, 412943252615.63153, 341375166.66254705],   # verified 2026-10-29 02 (logs/cron/2026-10-29-02_verify_dqi_quantum-circuits-22-75-local-gm-897-qubit-dqi.log); dqi theirs; chemistry kirilltym's gauge on heliosli v12's circuits; invsqrt busyaprime 22.12's, byte-identical
                        'Gluzdov 22.72': [102664.23957721348, 414781342770.7355, 340990159.68148416],   # verified 2026-10-29 02 (logs/cron/2026-10-29-02_verify_chem_quantum-circuit-compilation-from-source.log); chemistry his; invsqrt busyaprime 22.12's and dqi heliosli 22.54's, byte-identical (listed after them, so a tie goes to the author)
                        'heliosli 22.74': [102349.2360480486, 412943252615.63153, float('inf')],   # 2026-10-29 01:08 run; invsqrt busyaprime 22.72's and dqi somejoint 22.75's, byte-identical (listed after them, so a tie goes to the author); its own chemistry (their log: 3.41133e8) is not verified here, its circuits are used in the combination below
                        'heliosli circuits + Gluzdov gauge': [float('inf'), float('inf'), 340781040.24098283],   # chemistry only, combined in section 1 from the two above; verified 2026-10-29 02 (logs/cron/2026-10-29-02_verify_chem_combo_helio_gluzdov.log)
                        'Gluzdov 22.75': [102349.2360480486, 412943252615.63153, 340650168.69315076],   # verified 2026-10-29 08 (logs/cron/2026-10-29-08_verify_chem_gluzdov-f8a62242.log); chemistry his; invsqrt busyaprime 22.72's and dqi somejoint 22.75's, byte-identical (listed after them, so a tie goes to the author)
                        'heliosli 22.75': [102349.2360480486, 412943252615.63153, 340571920.8004815],   # verified 2026-10-29 08 (logs/cron/2026-10-29-08_verify_chem_heliosli-952da52c.log); chemistry theirs; invsqrt busyaprime 22.72's and dqi somejoint 22.75's, byte-identical
                        'heliosli 22.75 circuits + Gluzdov 22.75 gauge': [float('inf'), float('inf'), 340441129.56165034],   # chemistry only, combined in section 1 from the two above; verified 2026-10-29 08 (logs/cron/2026-10-29-08_verify_chem_atakan-b5d50e89.log); atakanaldemir's Iso-Gap ships the same bytes
                        'Atakan 888Q': [102349.2360480486, 412372251652.37823, 339728082.3378603],   # verified 2026-10-29 16 (logs/cron/2026-10-29-16_verify_dqi_atakan-736f3055.log, logs/cron/2026-10-29-16_verify_chem_d04595e4.log); dqi and chemistry his; invsqrt busyaprime 22.72's, byte-identical
                        'kirilltym 5-word QROM': [102349.2360480486, 410274633000.4907, 339728082.3378603],   # verified 2026-10-29 16 (logs/cron/2026-10-29-16_verify_dqi_kirilltym-7a378f47.log); dqi hengck23's dqi_iter014 (hengck23 ships no submission.zip); chemistry Atakan's and invsqrt busyaprime's, byte-identical (listed after Atakan, so a chemistry tie goes to the author)
                        'Özer 22.90': [100568.4991945597, 393389075235.9089, 331076721.7501548],   # verified 2026-10-31 20 (logs/cron/2026-10-31-20_verify_dqi_ahmedberatozer-909410f5.log, logs/cron/2026-10-31-20_verify_chem_ahmedberatozer-de159897.log); his v18 (2026-10-31 17:30 run): dqi SomeJoint's 900-qubit decoder with his 576 history caches (909410f5), chemistry his λ polish of Lera's C42 table (de159897), invsqrt Ehan Mallick's; until v70: [100940.56165415398, 403339761880.02136, 337798090.4640843]; verified 2026-10-30 20 (logs/cron/2026-10-30-20_verify_chem_manwithacat150k-b52f5d6d.log); his v16 (15:37 run): chemistry his new weights; invsqrt his v15 and dqi manwithacat Voodoo's, byte-identical; until v61: [100940.56165415398, 410274633000.4907, 338047050.9649815] (his 11:12 output)
                        'manwithacat Voodoo': [100940.56165415398, 403339761880.02136, 337833376.1326832],   # verified 2026-10-30 16 (logs/cron/2026-10-30-16_verify_dqi_manwithacat-2ed7ccf1.log, logs/cron/2026-10-30-16_verify_chem_manwithacat-84fcab7e.log); his VQCC Quantum Voodoo v4 (15:00 run): dqi his 888-qubit failure fix, chemistry his lambda refit; invsqrt Özer v15's, byte-identical (listed after it)
                        'manwithacat Real $150K': [100568.4991945597, 387059207747.3355, 327359144.12415844],   # verified 2026-11-01 08 (logs/cron/2026-11-01-08_verify_{chem_manwithacat-69da5111, dqi_heliosli-fb2da4f2}.log); his 2026-11-01 02:53 run: chemistry 69da5111 (lambda 47.1059, 16,237 Toffolis), dqi ahoirg's 893-qubit build (fb2da4f2, as heliosli's), invsqrt Ehan Mallick's, byte-identical; until v72: [100568.4991945597, 393944554828.0952, 331025488.42506784]; verified 2026-10-31 20 (logs/cron/2026-10-31-20_verify_dqi_somejoint-38f0ac17.log, logs/cron/2026-10-31-20_verify_chem_manwithacat150k-c64292ad.log); his 2026-10-31 17:29 run: dqi SomeJoint's 900-qubit decoder (38f0ac17), chemistry his Hedge Wizardry HW-04 (c64292ad), invsqrt Ehan Mallick's, byte-identical; until v70: [100692.58221354822, 402458290261.8694, 337798090.4640843]; his 18:42 output: yoyobar v8's invsqrt and dqi and Özer v16's chemistry, byte-identical (listed after them, so a tie goes to the authors); until v61, above Özer 22.90: [102349.2360480486, 410274633000.4907, 338130690.99867845] (his 07:47 output)
                        'manwithacat Hedge Wizardry': [100568.4991945597, 398268953060.8457, 331025488.42506784],   # verified 2026-10-31 20 (logs/cron/2026-10-31-20_verify_dqi_manwithacathw-fe98677c.log, logs/cron/2026-10-31-20_verify_chem_manwithacat150k-c64292ad.log); his 2026-10-31 16:29 run: chemistry his HW-04 (c64292ad, refit of Lera's C42 weights), dqi fe98677c, invsqrt Ehan Mallick's; until v70: [100692.58221354822, 402702373901.5095, 337651009.063524]; verified 2026-10-31 08 (logs/cron/2026-10-31-08_verify_chem_manwithacat-d4a8f109.log); output as mounted from 2026-10-31 (see EXTRA); until v64: [100692.58221354822, 401986129571.4192, 337798090.4640843]; verified 2026-10-31 02 (logs/cron/2026-10-31-02_verify_dqi_heliosli-5faba459.log, the same bytes from heliosli's output); his 22:06 output: dqi his HW-01 on yoyobar v8's decoder; invsqrt yoyobar v8's and chemistry Özer v16's, byte-identical (listed after them)
                        'Leo Tang T7421': [100568.4991945597, 387059207747.3355, 327359144.12415844],   # verified 2026-11-01 08 (logs/cron/2026-11-01-08_verify_{chem_manwithacat-69da5111, dqi_heliosli-fb2da4f2}.log); his 2026-11-01 05:28 run: byte-identical to manwithacat Real $150K's 2026-11-01 output (dqi fb2da4f2, chemistry 69da5111, invsqrt bce2948f); until v72: [100568.4991945597, 401986129571.4192, 333395100.914337]; verified 2026-10-31 20 (dqi HW-01 5faba459 as verified in logs/cron/2026-10-31-02_verify_dqi_heliosli-5faba459.log, logs/cron/2026-10-31-20_verify_chem_leotang09-e574ad75.log); his 2026-10-31 17:21 run: chemistry e574ad75, dqi Hedge Wizardry HW-01 (5faba459), invsqrt Ehan Mallick's, byte-identical; until v70: [100692.55756592617, 401986129571.4192, 337780864.9643602]; verified 2026-10-31 08 (logs/cron/2026-10-31-02_verify_inv_leotang09-2366b67c.log); output as mounted from 2026-10-31 (see EXTRA); until v64: [100692.55756592617, 401986129571.4192, 337798090.4640843]; verified 2026-10-31 02 (logs/cron/2026-10-31-02_verify_inv_leotang09-2366b67c.log); his 23:34 output: invsqrt yoyobar's candidate 1599 from yoyobar's dataset, unchanged; dqi Hedge Wizardry HW-01 and chemistry Özer v16's, byte-identical (listed after them)
                        'somejoint 23.77': [100568.4991945597, 385311255783.3984, 327300810.34268564],   # verified 2026-11-01 08 (logs/cron/2026-11-01-08_verify_{dqi_somejoint-bdee1848,chem_somejoint-f2088294}.log); their 2026-11-01 06:05 run: dqi their 892-qubit decoder with Ali Haydar Özdağ's (ahoirg) flag carry on yoyobar's one-carry history (bdee1848), chemistry heliosli's version-33 weights on Lera's (9, 143, 40) factorization (f2088294), invsqrt Ehan Mallick's, byte-identical
                        'heliosli 22.81': [100568.4991945597, 387059207747.3355, 327053613.2408534],   # verified 2026-11-01 08 (logs/cron/2026-11-01-08_verify_{chem_heliosli-681a8a8d, dqi_heliosli-fb2da4f2}.log); their 2026-11-01 05:40 run: chemistry Lera's (marculera) (9, 143, 40) factorization and angle circuits with their newer table weights (681a8a8d, lambda 47.0619, 16,237 Toffolis); dqi ahoirg's (Ali Haydar Özdağ) 893-qubit build of SomeJoint's decoder (fb2da4f2); invsqrt Ehan Mallick's, byte-identical; until v72: [100568.4991945597, 393389075235.9089, 328711157.2734612]; verified 2026-11-01 02 (logs/cron/2026-11-01-02_verify_chem_heliosli-4e1d0424.log); their 2026-10-31 23:40 run: chemistry Lera's (marculera) (9, 143, 40) factorization and angle circuits with their table weights (4e1d0424, lambda 47.2074, 16,269 Toffolis); dqi Özer v18's (909410f5) and invsqrt Ehan Mallick's, byte-identical; until v71: [100568.4991945597, 400273711353.48145, 337521609.31504345]; verified 2026-10-31 14 (logs/cron/2026-10-31-14_verify_chem_heliosli-bbbc7337.log); their 10:20 output (see EXTRA): chemistry theirs, invsqrt Ehan Mallick's and dqi yoyobar's 07:01, byte-identical; until v67: [100692.58221354822, 400273711353.48145, 337566987.22221684]; verified 2026-10-31 09 (logs/cron/2026-10-31-09_verify_chem_heliosli-90f3e877.log); their 08:08 output (see EXTRA); until v66: [100692.58221354822, 402702373901.5095, 337779790.59433043]; verified 2026-10-31 08 (logs/cron/2026-10-31-08_verify_chem_heliosli-3821c71a.log); output as mounted from 2026-10-31 (see EXTRA); until v64: [100692.58221354822, 401986129571.4192, 337780864.9643602]; verified 2026-10-31 02 (logs/cron/2026-10-31-02_verify_chem_heliosli-b684dc6f.log); their 00:05 output: chemistry their new weights; invsqrt yoyobar v8's and dqi Hedge Wizardry HW-01, byte-identical (listed after them, so a tie goes to the authors); until v62, below Gluzdov 22.81: [100940.56165415398, 410274633000.4907, 337991954.0482345] (their 13:43 output)
                        'somejoint 22.77': [102349.2360480486, 414211737069.8777, 340232090.43014985],   # verified 2026-10-29 14 (logs/cron/2026-10-29-14_verify_dqi_ozer-6349d2bc.log, logs/cron/2026-10-29-14_verify_chem_somejoint-b143fe4b.log); dqi and chemistry theirs; invsqrt busyaprime 22.72's, byte-identical
                        'Gluzdov 22.81': [100692.58221354822, 398941122234.2813, 337714978.8792064],   # verified 2026-10-31 20 (logs/cron/2026-10-31-20_verify_dqi_dmitriigluzdov-32e30c08.log, logs/cron/2026-10-31-20_verify_chem_dmitriigluzdov-3eca89fc.log); his 2026-10-31 15:48 run: chemistry 3eca89fc, dqi 32e30c08, invsqrt yoyobar v8's (67482db6); until v70: [101002.47843760978, 410274633000.4907, 338130690.99867845]; verified 2026-10-30 08 (logs/cron/2026-10-30-08_verify_chem_yoyobar_gluzdov-9c93ad9b.log); his 04:39 output: chemistry his v44 (lambda 48.343); invsqrt yoyobar's and dqi hengck23's, byte-identical; until v58: [102349.2360480486, 410274633000.4907, 338879606.6099476] (his 00:02 output)
                        'heliosli 22.81 circuits + Gluzdov 22.81 gauge': [float('inf'), float('inf'), 339861567.46676314],   # chemistry only, combined in section 1 from the two above; verified 2026-10-29 14 (logs/cron/2026-10-29-14_verify_chem_combo-helio-gluzdov.log)
                        'somejoint 22.67': [102664.23957721348, 414781342770.7355, 343172572.78225887],   # verified 2026-10-28 14 (logs/cron/2026-10-28-14_verify_chem2_quantum-circuits-22-67-local-gm-full-xor-chem.log); chemistry theirs; invsqrt busyaprime's and dqi heliosli 22.54's, byte-identical (listed after them, so a tie goes to the author)
                        'atakanaldemir 22.66': [102664.23957721348, 414781342770.7355, 343787901.22637963],   # verified 2026-10-28 14 (logs/cron/2026-10-28-14_verify_chem_quantum-circuits-c60-reachable-cache.log); chemistry: somejoint 22.63's factorization with his angle loaders; invsqrt busyaprime's and dqi heliosli 22.54's, byte-identical (listed after them, so a tie goes to the author)
                        'kirilltym 22.64': [102664.23957721348, 414781342770.7355, 344812530.9004197],   # verified 2026-10-28 14 (logs/cron/2026-10-28-14_verify_chem_quantum-circuits-exact-gauge-5-word-qrom.log); chemistry: somejoint 22.63's circuits with his exact gauge; invsqrt and dqi byte-identical to the above
                        'atakanaldemir circuits + kirilltym gauge': [float('inf'), float('inf'), 343477342.01019543],   # chemistry only, combined in section 1 from the two above; verified 2026-10-28 14 (logs/cron/2026-10-28-14_verify_chem_combo_atakan_kirill.log)
                        'somejoint 22.34': [103546.168164559, 422346240055.8793, 349339717.3328993],   # verified 2026-10-28 (logs/cron/2026-10-28-02_verify_dqi.log); dqi theirs; invsqrt mrmeows 21.0's and chemistry heliosli 22.13's, byte-identical (listed after them, so a tie goes to the author)
                        'heliosli J5': [float('inf'), 448601212111.76025, 376433626.307489],   # its 2026-10-27 08:00 version, verified here: chemistry heliosli's; dqi somejoint's 21.34 fused decoder, byte-identical (invsqrt is mrmeows 21.0's)
                        'somejoint 21.20': [103546.168164559, 450382244807.3486, 383267019.2215811],   # verified 2026-10-27; its invsqrt is mrmeows 21.0's and its chemistry heliosli's, byte-identical
                        'Rustad C46': [103546.168164559, 455954457807.33057, 369005050.67107165],   # verified 2026-10-27 (logs/gauge_rustad_c46.log); chemistry his R10 C46 (423 Q); invsqrt mrmeows 21.0's and dqi somejoint 20.96's, byte-identical
                        'busyaprime 21.88': [float('inf'), float('inf'), 355065066.0799456],   # verified 2026-10-27 (chemistry only here: its invsqrt and dqi are byte-identical to Özer v9's, credited there)
                        'Özer v9': [103546.168164559, 444151956391.4917, 355186805.37080276],   # verified 2026-10-27 (logs/verify_ozer_1027e.log); dqi and chemistry his; invsqrt mrmeows 21.0's, byte-identical
                        'Kirill':   [182600.8579943692, 2574955459088.485, 2529044626.5199904],
                        'magmacot': [129833.62311348258, 891688391401.3619, 543925450.5623397],
                        'leoprovorov': [114323.93229681085, 842819878125.7235, 514410995.12465084],  # mrmeows's invsqrt and dqi, Özer's chemistry
                        'Özer':     [float('inf'), float('inf'), 494962019.3347939],   # his 2026-10-26 evening version; its invsqrt (mrmeows's) and dqi (our 883-qubit Schur) are byte-identical copies, credited to their sources
                        'Gluzdov':  [120219.23090505149, 891688391401.3619, 513261654.6882092],
                        'mrmeows open circuits': [108560.35824641786, 8.008e11, 501480157.9635119],   # dqi as stated in its README, not verified here (ours is cheaper)
                        'rustad + mrmeows compiler (ours)': [float('inf'), float('inf'), 436969552.20496726],   # chemistry only: rustad's (10,129,60) words, mrmeows's compiler, J 3
                        'rustad + mrmeows compiler, 4 words (ours)': [float('inf'), float('inf'), 425969844.4690444],   # J 4: four cached monomials erased mid-circuit and recomputed into the freed wires, 128 angle qubits
                        'Özer v8 (Bézout middle block)': [float('inf'), 495048250576.62756, float('inf')],   # dqi only: our 883-qubit Schur build, its amplitude conversion replaced by mrmeows's round17 Bézout transform, plus the Toffoli-free mask projection
                        'Özer v8 at 882 qubits (ours)': [float('inf'), 485193743459.73724, float('inf')],   # dqi only: Özer v8 with our 882-qubit reverse synthesis (fan-out through the term's first target, no scratch wire)
                        'Schur (jday96314)': [114323.93229681085, 500874373991.25, 508908870.7302555]},   # dqi rebuilt from jday96314/QuantumAI, our 883-qubit reverse synthesis
                       index=['invsqrt', 'dqi', 'dfthcsos'])
display(OFFLINE.style.format('{:.6g}').highlight_min(axis=1))
# so only the cheapest circuit per task is re-verified here (one verifier run per task instead of one per candidate)
PICK = OFFLINE.idxmin(axis=1).to_dict()
DIRS = dict(EXTRA_DIRS, Kirill=PUB)
res, SRC = {}, {}
for t, name in PICK.items():
    j = verify(t, DIRS[name]) if name in DIRS else {'status': 'NOT_ATTACHED'}
    if j.get('status') != 'PASS':                            # the pick failed here: fall back to every other candidate
        print(f'{t}: {name} {j.get("status")}; trying the others')
        for other in OFFLINE.loc[t].sort_values().index.drop(name):
            if other in DIRS and (j := verify(t, DIRS[other])).get('status') == 'PASS':
                name = other; break
    res[t], SRC[t] = j, DIRS.get(name)
    print(f"{t}: {name} {j.get('status')} {j.get('score')} ({j['sec']} s)")
# ours: the LLM-iterated gauge rotation of that chemistry's factor (section 4b), computed offline and shipped in the
# attached dataset as gauged_<sha of the public submission.npz>.npz; here it is applied and verified like any candidate.
# The angle words and the 25 circuits are unchanged; only W and H2_I in submission.npz are rotated (Gram kept).
import numpy as np
art = Path(SRC['dfthcsos']) / 'dfthcsos'
sha = hashlib.sha256((art / 'submission.npz').read_bytes()).hexdigest()[:12]
G = next(iter(glob.glob(f'/kaggle/input/**/gauged_{sha}.npz', recursive=True)), None) if KAGGLE \
    else next(iter(glob.glob(f'../out/gauge_ship/gauged_{sha}.npz')), None)
print('gauge for this chemistry:', G or f'none shipped for {sha}')
if G:
    OURS_DFT = OUT / 'ours_dft'; shutil.rmtree(OURS_DFT, ignore_errors=True); shutil.copytree(art, OURS_DFT / 'dfthcsos')
    with np.load(art / 'submission.npz', allow_pickle=False) as z:
        fac = {k: np.array(z[k]) for k in z.files}
    with np.load(G, allow_pickle=False) as z:
        W1, H1 = np.array(z['W']), np.array(z['H2_I'])
    a0 = np.concatenate([fac['W'], fac['H2_I'][:, None, :]], axis=1); a1 = np.concatenate([W1, H1[:, None, :]], axis=1)
    drift = float(np.max(np.abs(a1 @ a1.swapaxes(1, 2) - a0 @ a0.swapaxes(1, 2))))
    l1 = lambda m: float(np.sum(np.abs(m).sum(axis=1) ** 2))
    print(f"gauge: column L1 sum {l1(a0):.3f} -> {l1(a1):.3f}, Gram drift {drift:.1e}")
    assert drift < 1e-8
    np.savez(OURS_DFT / 'dfthcsos' / 'submission.npz', **dict(fac, W=W1, H2_I=H1))
    j = verify('dfthcsos', OURS_DFT)
    print(f"dfthcsos ours ({PICK['dfthcsos']} + LLM gauge): {j.get('status')} {j.get('score')} lambda {j.get('lambda_eff')} vs {res['dfthcsos'].get('score')}")
    if j.get('status') == 'PASS' and j['score'] < res['dfthcsos'].get('score', float('inf')):
        res['dfthcsos'], SRC['dfthcsos'] = j, OURS_DFT
tab = pd.DataFrame([(t, j.get('status'), j.get('num_qubits'), j.get('score'), REF_COST[t] / j['score'] if j.get('status') == 'PASS' else None, j['sec'])
                    for t, j in res.items()], columns=['task', 'verifier', 'qubits', 'cost', 'ratio', 'seconds'])
display(tab)
ok = tab.ratio.notna().all()
gm = math.prod(tab.ratio) ** (1 / 3) if ok else float('nan')
print(f'geometric mean of the ratios: {gm:.4f}' + ('' if ok else '  (a component failed: not submitted)'))
```

```python
if ok:
    with zipfile.ZipFile(OUT / 'submission.zip', 'w', zipfile.ZIP_DEFLATED) as z:
        for t in ('invsqrt', 'dqi'):
            p = find(SRC[t], t); z.write(p, p.name)
        for p in sorted((SRC['dfthcsos'] / 'dfthcsos').iterdir()):
            z.write(p, f'dfthcsos/{p.name}')
    print('wrote', OUT / 'submission.zip', f"{os.path.getsize(OUT / 'submission.zip') / 1e6:.1f} MB")
```

## 3. The Schur dqi: whose it is, and what we added

**The decoder is jday96314's.** It is the raw continued-fraction Schur decoder from their public GitHub repository
[jday96314/QuantumAI](https://github.com/jday96314/QuantumAI) (round 6, `initial_experiments/6/agent14`): quadratic-extension Schur extraction,
continued-fraction synthesis `(N, B) ← (l_i N + q_i B, N)`, an amplitude polynomial `C = A / B′ mod B` so the scan costs 11 Toffolis per point,
a three-field endpoint and a remainder-consuming quotient. Their own write-up explains each step; all credit for the design goes to them.

**What we did.**

1. *Rebuilt it from their code.* Their 16-AND cubic cover file is not in the repository; we regenerated it with their own search, which found 17 ANDs,
   so our build has 535,334 Toffolis against their 534,575 at the same 892 qubits. The official verifier passes it at 5.1516 × 10¹¹.
2. *Took 9 qubits off its peak (892 → 883).* The cost is repetitions × qubits² × Toffolis, so width counts twice.
   Tracing every allocation showed the 892 peak sits in one place only: the reverse synthesis's destructive quotient, at the largest fold step,
   where all 876 port wires are live and the quotient's private workspace reaches 16 wires (the tower-inverse chain, and the trace-inverse phase kernel).
   Three changes, each tested on their small-N checks (values, scratch, phase) before the full build:
   * the κ·n₁·d₁ product is written straight into its output, so a 12-wire window disappears (same 42 CCX);
   * the 12 output wires of the scalar inverse are provably zero while the tower chain and the phase kernel run, so those kernels borrow them (16 → 7 and 16 → 4 private wires);
   * the norm is measured away during the two scalings, its 12 wires serve as their temporary, and it is recomputed afterwards (+42 Toffolis per quotient); the held measurement phases are repaid with conditioned CZs at no Toffoli cost.

   Result: 883 qubits and 536,804 Toffolis (+0.27 %), official verifier **PASS at 5.0087 × 10¹¹** (1 value error, 0 phase errors in 102,400 shots).
   The width change itself is worth (883/892)² × (536,804/535,334) = **−1.74 %**; the rest of the measured −2.77 % is the sampled value-error count (1 here, 4 before), which we count as luck.
3. *Where it stops.* 883 is also the peak of the middle stages, so going lower needs those cut together with the reverse synthesis.
   The reverse extraction costs 31 Toffolis more per node pair than the forward one (595 pairs: 132,685 against 114,240):
   the reverse has to repay the phase of its own norm measurement, and every route we found costs the same 30 Toffolis, so we left it.


## 4. An LLM-iterated gauge for the chemistry factor (how we did it)

The chemistry cost is roughly λ × Toffolis. Each slice of the factor [W; H₂] may be rotated on the right by any orthogonal
matrix: the Gram matrix, so the Hamiltonian, stays the same, and so do the angle words and the 25 angle-loading circuits.
A rotation that lowers the column L1 sum lowers λ, and with it the cost. Section 2 applies such a rotation to the best
public chemistry (the cheapest one that verifies) and verifies it like every other candidate.

**How we got the rotation code.** The idea, an LLM iterating on the generator code against the official verifier
(as in ShinkaEvolve or AlphaEvolve), is Alekh's. Our implementation: a task file stating the goal, the one function that may
change and the invariants, plus a scorer that builds the circuits with a candidate and runs the competition's own
dfthcsos verifier (about 80 s per candidate). The agent was Gemini in Google's Antigravity IDE: it edited, scored and kept
its best in a loop. The same task with island evolution (ShinkaEvolve) and a local 30B model found nothing in 15 generations.

Gemini replaced the Cayley-transform descent with Riemannian conjugate gradient (Polak–Ribière with restarts), longer
stages with cached losses, periodic SVD re-orthogonalisation and Givens coordinate-descent sweeps.
Two further rounds (on Özer's factor, then on Artem's newest) added more stages, multiscale Givens sweeps and seeded orthogonal basin-hopping (L1 sum on Artem's factor 1120.34 → 1116.03).

| factor | public cost | with the LLM gauge | change |
|---|---|---|---|
| Magmucot (65 × 60 × 30) | 2.0394e9 | 2.0344e9 (first try) | −0.25 % |
| Özer (32 × 60 × 30) | 1.19209e9 | 1.18600e9 | −0.51 % |
| Artem, Compressed Group Compiler (same 36,861 Toffolis) | 1.17492e9 | 1.16888e9 (λ 48.44 → 48.19) | −0.51 % |
| Artem, Progressive Group Compression (same 36,022 Toffolis) | 1.12815e9 | 1.12606e9 | −0.19 % |
| Özer, evening version (26 slices, same 31,910 Toffolis) | 1.02167e9 | 1.01610e9 (λ 48.54) | −0.55 % |
| DePie, 2026-10-25 version (26 slices, same 31,860 Toffolis) | 9.97913e8 | 9.93196e8 (λ 47.52) | −0.47 % |
| Özer, 2026-10-25 version (26 slices, same 30,266 Toffolis) | 9.63250e8 | 9.60045e8 (λ 48.35) | −0.33 % |
| Özer v4, Gluzdov's (1, 496, 384) factor (same 20,286 Toffolis, 833 qubits) | 8.21434e8 | 8.17988e8 (λ 48.41) | −0.42 % |
| magmacot 76Q-881Q blend, 448-qubit chemistry with offset polish (same 25,394 Toffolis, 448 qubits) | 5.45371e8 | 5.43925e8 (λ 47.94 → 47.81, v18 on Kaggle) | −0.27 % |

The optimiser is in the attached dataset `alexchilton/qcc-llm-gauge` (`gemini_gauge.py`). Base code: Magmucot's
*Clean Workspace Reuse* gauge generator (Apache-2.0); the circuits it is applied to: Artem1410 and Ahmed Berat Özer (Apache-2.0).

## 5. Where the points are (measured 2026-10-26)

* **dqi (Schur).** 883 qubits is now also the peak of the decoder's middle stages, so going narrower needs those and the reverse synthesis cut together.
  The reverse extraction costs 31 Toffolis more per node pair than the forward one (595 pairs: 132,685 against 114,240).
  That extra is the phase repayment for its own norm measurement, and every route we found costs the same.
  Every stage runs on the same 42-AND GF(2¹²) multiplier, so a cheaper field multiplier would pay everywhere; jday96314's searches did not find one.
* **dfthcsos.** Of the 440 to 448 qubits, 321 are the cost model's temporary workspace for the inner state preparation and 87 are its persistent qubits.
  The angle circuits' own 77 temporaries are below that maximum, so making them narrower saves nothing.
  The levers are the model's inputs: the factor's shape and the word width brot.
  mrmeows's refit gained from a discontinuity in Q; the brot ≤ 7 cliff (section 6c) would halve the temporaries if a refit still meets the 20 mHa gates.
* **invsqrt.** In the 63-qubit circuit all 63 wires hold nonzero values for a third of the run, mostly in the 32 Horner rows, with no linear relation among them.
  The measure-instead-of-uncompute trick is already used (3 of 1,250 ANDs are still uncomputed by gates), and the 44 rotations have no slack.
  One qubit is worth about 1.5 % on invsqrt, but it has to come from different row arithmetic, not from a gate-level patch.

## 6. Earlier work: cost anatomy of the 2026-10-23 portfolio

### 6a. invsqrt: the reference spends 98 % computing a number and erasing it

The reference follows the paper: normalise x, look up a cubic initial guess, one Newton–Raphson step, phase by the bits
of the result, then undo everything. The table below reads the flame chart. The phase itself, the only part that
produces the output, is a sliver; computing y₁ and uncomputing it are almost all of the cost.

```python
ref_i, pub_i = anatomy(find(REF, 'invsqrt')), anatomy(find(PUB, 'invsqrt'))
print(f"reference: {ref_i['qubits']} qubits, {ref_i['toffoli']} Toffolis | best public: {pub_i['qubits']} qubits, {pub_i['toffoli']} Toffolis")
display(pd.concat([top_table(ref_i, 'reference'), top_table(pub_i, 'best public')]).set_index(['circuit', 'stage']))
display(Markdown('**Reference flame chart** (red = stages, nested upward; blue = live qubits):')); display(SVG(ref_i['svg']))
display(Markdown('**Best public flame chart:**')); display(SVG(pub_i['svg']))
```

The public design (Sumit Chongder's, refined by anazemcev) does not tune Newton–Raphson; it **removes it**. After a binary
normalisation it evaluates a quadratic directly, in centred signed-digit Horner form, with the coefficients for each
interval fetched by a table lookup (the tagged `lookup` / `unlookup` stages); the Horner arithmetic is the untagged part.
One polynomial instead of a cubic guess plus a Newton step: about 95 qubits instead of 260, about 1,500 Toffolis instead of 7,735.
Where to look next: the untagged Horner arithmetic, which is now most of the cost.

### 6b. dqi: Euclid out, Berlekamp–Massey in

The reference decoder runs the extended Euclidean algorithm (EEA) forward to find the error locator, a Chien search +
Forney step to fix the sign, and the EEA in reverse to clean up. The best public decoder replaces the EEA with
Berlekamp–Massey.

```python
ref_d, pub_d = anatomy(find(REF, 'dqi')), anatomy(find(PUB, 'dqi'))
print(f"reference: {ref_d['qubits']} qubits, {ref_d['toffoli']:,} Toffolis | best public: {pub_d['qubits']} qubits, {pub_d['toffoli']:,} Toffolis")
display(pd.concat([top_table(ref_d, 'reference'), top_table(pub_d, 'best public')]).set_index(['circuit', 'stage']))
display(Markdown('**Best public flame chart:**')); display(SVG(pub_d['svg']))
```

The score is R(ε) · Q² · T, so **qubits count twice**: a 1 % cut in qubits is worth a 2 % cut in Toffolis.
The best public decoder cut Toffolis about 10× but qubits only 4 % (1,893 → 1,818). Where its Toffolis go (measured on
Kirill's 2026-10-23 morning version, 583,736 Toffolis; his current dqi costs 0.5 % less):

| stage | Toffolis | share | per unit |
|---|---|---|---|
| Berlekamp–Massey forward (70 iterations) | 148,004 | 25.4 % | 2,114 per iteration |
| monicise + Ω + restore + unmonicise | 15,420 | 2.6 % | |
| sweep: one Forney step per position × 4,095 | 327,600 | 56.1 % | exactly 80 per position |
| Berlekamp–Massey reverse | 92,712 | 15.9 % | |

**The 80 Toffolis of one Forney step**, read from the circuit text: 11 for the zero test (an AND chain over the 12 bits
of σ(α⁻ʲ)), then 69 for one GF(2¹²) inversion of σ′(α⁻ʲ) done as a tower over GF(64) and GF(8) (norm 15 + 6 + 3,
conditioning 3, two sub-field products 12 + 30). The phase kickback of ⟨Ω·inv, M_j⟩ is CZ-only and free, and the
inverse is measured away. The reference pays 11 + 351 per position; a tower inversion of GF(2¹²) needs about 65 ANDs,
so this step is at its floor. Live qubits sit at 96–100 % of 1,818 in every stage, so there is no idle window either:
the next dqi gain needs a different decoder, not a better gate. Section 3b brings one in from the literature.

### 6c. dfthcsos: the weakest task in public

Here the circuits are a family of `ANGLE_k` QROM lookups that load Givens-rotation angles for a factorised FeMoCo
Hamiltonian; the rest of the block encoding is priced by formula. Most of the score lives in the classical
factorisation (λ_eff) and in how many angle bits, and how many angles per batch, the circuits load.

```python
def family(folder):
    rows = []
    for p in sorted(Path(folder, 'dfthcsos').glob('angle_k_*.km?'), key=lambda p: int(re.search(r'(\d+)', p.stem).group(1))):
        c = km.Circuit.from_file(str(p)); rows.append((p.stem, c.num_qubits, c.operation_statistics().num_toffoli_like))
    return pd.DataFrame(rows, columns=['circuit', 'qubits', 'Toffolis'])
fr, fp = family(REF), family(PUB)
print(f"reference: {len(fr)} ANGLE circuits, {fr.Toffolis.sum():,} Toffolis, widest {fr.qubits.max()} qubits")
print(f"best public: {len(fp)} ANGLE circuits, {fp.Toffolis.sum():,} Toffolis, widest {fp.qubits.max()} qubits")
pd.concat([fr.describe().loc[['mean', 'min', 'max']].assign(set='reference'), fp.describe().loc[['mean', 'min', 'max']].assign(set='best public')])
```

Public notebooks have moved this task only about 1.43×, against roughly 10× on the other two, so under the
geometric mean it is where a point of improvement is worth the most.

**The cost model has a cliff.** The verifier prices dfthcsos as λ_eff × T × Q, where T is twice the ANGLE Toffolis
plus a formula for the rest of the block encoding, and Q includes the formula's temporary qubits
(`verifiers/dfthcsos/SCORING_README.md`, `scoreinfra/costs_prepare.py`). That formula keeps its inner QROAM workspace
below 75 × brot qubits, so at **brot ≤ 7** (angle bits) it must pick a smaller block and its temporary qubits halve:

| brot | formula temp qubits | formula Toffolis |
|---|---|---|
| 8 … 12 | 543 | 8,194 … 9,394 |
| 4 … 7 | 272 | 10,622 … 11,522 |

What the cost model alone says at the current λ_eff (the 20 mHa accuracy gates are **not** checked here):

| shape (R, B, C) | brot | T | Q | ratio |
|---|---|---|---|---|
| (65, 60, 30), best public | 10 | 72,514 | 681 | 1.43 |
| (65, 60, 30) | 7 | 58,206 | 409 | 2.96 |
| (65, 32, 30) | 10 | 43,877 | 649 | 2.48 |

**Why nobody has taken it: rounding does not get there.** Rounding the best public angle words to fewer bits, checked
with the verifier's own `evaluate_words`: error₂ goes from 19.98 mHa (the ceiling is 20) to 188 at brot 9 and 341 at
brot 7; one least-significant bit on every two-body word alone costs +112 mHa. The weights were co-fitted to these exact
words, so a lower brot or batch needs a quantization-aware refit of the factorisation (the starter kit's JAX trainer),
not post-hoc rounding. That is the one place in this competition with a 2× ceiling we can see. wanxiaolong's measurements (table above) show 7-bit two-body QAT stuck at error₂ ≈ 0.11 with every optimizer tried, and the same optimizer cannot recover a known 8-bit solution, so the 7-bit question is still open: the search is the bottleneck, not yet the bits.

## 7. Earlier work: our invsqrt v6

**The design everyone builds on** is [Sumit Chongder's](https://www.kaggle.com/code/sumitchongder/notebook2f196e646b):
binary normalisation, one 256-entry QROM of quadratic coefficients, centred signed-digit Horner with Gidney adders,
one `Z_POW` per output bit, mirrored uncompute (103 qubits). Since then, in public:

* [anazemcev / Artem1410](https://www.kaggle.com/code/anazemcev/quantum-circuits-signed-digit-memory): phase and release the
  constant high word before the arithmetic, fold the last carry, borrow the always-1 bit `x[21]`: **93 qubits**, 175,203.
* [Rudra Bhaskar](https://www.kaggle.com/code/rudrabhaskar/highest-public-score-direct-polynomial-phasing): split the
  derivative accumulator and reuse the top carry: **95 qubits** but far fewer Toffolis.
* [Kirill](https://www.kaggle.com/code/kirilltym/quantum-circuits-qrom-tower42) combined Rudra's arithmetic with our
  changes 1 and 2 below: 95 qubits, 1,472 Toffolis, **168,521**. That is the best public circuit this notebook starts from.

The cost is qubits × (Toffolis + T/4), so one qubit is worth about 1 % and one Toffoli about 0.05 %.

**Our three changes.** We found them on Sumit's circuit (our v3) and measured them again on anazemcev's:

| | change | on anazemcev's circuit | Δ |
|---|---|---|---|
| | as published | 93 q, 1,582 T, 175,203 | |
| 1 | **Moves into empty qubits are not swaps.** When the barrel shifter moves a bit into a qubit that is still \|0⟩ for every shift, the controlled swap (1 Toffoli) becomes a temporary AND plus a CNOT, and the AND is undone by measurement on the way back. | 93 q, 1,562 T, 173,343 | −1.06 % |
| 2 | **Shared flags.** The normalisation needs three "are the top k bits all zero?" flags; one prefix-AND chain on the unshifted x gives all three, instead of three separate AND trees. | 93 q, 1,556 T, 172,785 | −0.32 % |
| 3 | **One bit less for the quadratic coefficient** (`F2` 10 → 9). The model's worst error rises from 0.928 to 0.945 of the tolerance, still inside, and the coefficient register, which is live at the peak, is one qubit shorter. | **92 q**, 1,554 T, **170,979** | −1.05 % |

**v6 = Kirill's current circuit + change 3 + a borrowed carry.**

| | step | qubits | Toffolis | cost (verifier) | Δ |
|---|---|---|---|---|---|
| | Kirill's circuit (Rudra's arithmetic + our changes 1–2) | 95 | 1,472 | 168,521 | |
| 3 | `F2` 10 → 9. His builder fixes the coefficient register at 9 qubits; we take the width from the configuration and handle the one R row whose shift turns negative at `F2` 9 by adding into a shifted target (exact, as in Sumit's original rows). At `F2` 10 our copy rebuilds his circuit **byte for byte**. | 94 | 1,470 | 166,800 | −1.02 % |
| 4 | **Borrow `x[21]` in Rudra's adders.** After normalisation x's top bit is always 1, and the arithmetic never reads it; anazemcev flips it to 0 and uses it as a clean carry wire in their adders. Rudra's adders did not, so we pass it to every one of them. | **93** | 1,470 | **165,026** | −1.06 % |

A live-qubit profile of v6 shows the 93-qubit peak is now reached in two places, the y rows and the high-word phase,
so the next qubit needs both to shrink; the only idle register at the y rows is the 8-bit coefficient c2, and
releasing it costs about +144 Toffolis (+8 %) for 1 qubit. This is close to the floor of the design.

Every change keeps the integer pipeline identical to the model (checked on all 3,133,452 inputs), so the arithmetic
certificate still covers it.

> **Pitfall that cost us a submission.** `Circuit.to_file(path)` writes the *text* format (`kmx`) unless you pass
> `format='kmb'`, whatever the file is called. The standalone verifier detects the format and passes the file; the
> leaderboard metric takes the format from the suffix and rejects a text file named `.kmb`, so the whole submission
> errors with no message. Write `c.to_file('invsqrt.kmb', format='kmb')`, or name text files `.kmx`.

**What we tried that did not pay** (all measured or counted, so you can skip them): our v3 trick of not storing y's
always-1 top bit also works here (the high word's bit 7 is 1 in all 256 entries), but that register is freed before the
peak, so it saves nothing; every other one-bit precision change fails the tolerance (on Sumit's circuit; the grid on this one is in the repo); cubic polynomials with 32 segments save 207 lookup Toffolis but cost ~450 more in the rows;
256 or 64 segments lose; a phase-gradient register needs the same 33 rotations plus 30 qubits; spending more error

## 8. Earlier work: dqi: a published decoder, ported to this task

**The decoder is not ours.** Gu, Kemeny, Yelin & Lukin, *Uncomputation …*
([arXiv:2610.22367](https://arxiv.org/abs/2610.22367)), publish a Reed–Solomon DQI decoder with an open
compiler ([andigu/uncompute](https://github.com/andigu/uncompute), Apache-2.0,
`tasks/dqi-reed-solomon/circuits/discovered`). In their words: a truncated Berlekamp–Massey decoder (70 shared-path
locator updates, each touching only the coefficient window that can already be active), Forney's formula contracted
with the phase mask so the decoded word is never formed, and a half-degree Euclidean pass that builds Ψ = 1/σ′ mod σ
once per shot, so the sweep pays only a root test and one gated pairing per point.
Their circuit: **1,837 qubits, 442,397 Toffolis**. It cannot be submitted as published, for three reasons:

| | their task | this competition | what the port does |
|---|---|---|---|
| mask | one fixed public mask, compiled into the gates | a new mask M_j per shot, in classical registers E0 … E4094 | the mask enters only through GF(2)-linear row masks, so every mask-gated CX becomes a fixed list of CXs, each conditioned on one mask bit. kickmix bills a bit-conditioned CX at 0 Toffolis. |
| fixed-mask shortcuts | 71 points removed by interpolation (a Z step), one-hot rows skipped | the mask is unknown at build time | every point is live, the Z step is dropped, the pairing always builds its row: +71 × 23 = **+1,633 Toffolis** |
| field | GF(2)[x]/(x¹²+x⁷+x⁶+x⁵+x³+x+1), generator γ = x | GF(2)[x]/(x¹²+x⁶+x⁴+x+1), generator α = x | φ(α) = γ⁴⁹⁹ is a field isomorphism; φ is applied in place to every syndrome word (CNOTs only), position j becomes 499·j mod 4095, and the mask is carried through φ⁻ᵀ |

Their op stream is then translated gate by gate into a kickmix circuit (`ops2kickmix.py`, written from the op-stream
spec in `uncompute/ops.py`). The truncation is a design choice of theirs: a rare input activates a coefficient beyond its
cutoff and gets the wrong phase, at about 1.4 × 10⁻⁴ per shot by their count. The verifier prices that through R(ε).

| circuit | qubits | Toffolis | R(ε) | cost R · Q² · T |
|---|---|---|---|---|
| reference | 1,893 | 5,725,108 | 1.191 | 2.44 × 10¹³ |
| best public (Kirill's portfolio) | 1,818 | 583,505 | 1.191 | 2.30 × 10¹² |
| this port, verified locally | 1,837 | 444,030 | 1.232 | 1.85 × 10¹² (−20 %) |
| this port + Kozykorpesh's four gadgets (built below) | **1,789** | 445,410 | 1.209 | **1.72 × 10¹²** (−25 %) |

**The four gadgets are Kozykorpesh's**, from his notebook
[*Quantum AI LB 6.68108*](https://www.kaggle.com/code/kozykappa/quantum-ai-lb-score-6-68108) (`gen_dqi_v6`, Apache-2.0):
`round2_gadgets.install` with `top_first`, `pebble_last` and `norm_pebble` (in his module's words: "the quotient swap
schedule, early Euclid quotient parking, and early linear root erasure") and `borrow_template` (his affine routing of two
B-bank words, so 24 qubits are lent instead of allocated). His module is copied unchanged; `build_gadgets.py` adds the two hooks they need to our port.
Per gadget on this port: borrow −24 qubits, top-first −18, each pebble −12; all four together −48 (1,837 → 1,789),
for +1,380 Toffolis (+0.3 %). On Q² · T that is −5 %.
The build below runs their compiler on the edited source, then the translation; it takes about 15 minutes and
about 3 GB of RAM.

```python
# The port is built from the attached source (src/dqi_port in alexchilton/uncompute-dqi-port) with:
#   make_dyn.py (their circuit + our edits) -> build_gadgets.py (+ Kozykorpesh's four gadgets) -> ops2kickmix.py
# That build takes about 27 minutes on Kaggle's CPU, so this notebook shows the result of its v24 run (2026-10-26)
# instead of rebuilding it on every version:
print("norm certificate: inverse correct on all 4096 GF(4096) elements (exact Gu term lists)")
print("emitted dqi_port.ops (204 s); expanded 132,362,017 ops, 445,410 CCX/CCZ, 1,789 qubits (410 s)")
print("wrote dqi_port/dqi.kmb, 1.24 GB (1,199 s)")
```
